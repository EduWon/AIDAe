## <span style="color: blue;">**2. Rationale for Weight Settings**</span>


## **Claude Response**

Good question. Let me explain the rationale for each factor's weight.

**Near order book imbalance (25%) — Highest weight**

The bid-ask volume difference in the 3 price levels immediately above/below the current price has the most direct impact on ultra-short-term price direction. Orders at distant price levels are frequently canceled or modified, but near-price orders are more likely to result in actual execution. In the uploaded data, the difference between the ask volume just above the current price (161,100: 15,800) and the bid volume just below (160,900: 11,965) suggests short-term directionality.

**Price momentum (20%)**

Looking at the 1-minute chart, the price surged from 158,800 at 18:13 to 161,600 at 19:08 and then consolidated. Whether this momentum continues or dissipates is a key variable determining the next move, so it was set as the second highest.

**Overall order book imbalance (15%) & Spread pressure (15%)**

The overall order book imbalance (ask 254K vs bid 159K) shows the big picture of supply and demand, but distant price levels contain many phantom orders, making them less reliable than near-price levels. Spread pressure serves as a supplementary confirmation signal, so it was set equally at 15%.

**Volume signal (15%)**

Following the principle that "price is confirmed by volume," this examines whether volume is supporting bullish candles. However, at the 1-minute candle level, noise is significant, so it was set lower than momentum.

**RSI signal (10%) — Lowest weight**

RSI is a lagging indicator, calculated on a 14-period basis, making it slow to react on 1-minute candles. It provides meaningful contrarian signals only in extreme overbought/oversold zones, so it was assigned a supporting role at 10%.

---

Honestly, these weights are initial values reflecting general conventions from market microstructure research. For precision use, it would be much more accurate to run backtests on historical data to find optimal weights, or to train a model like logistic regression/XGBoost. If you'd like, I can modify the dashboard to let you adjust weights via sliders in the UI.


## <span style="color: blue;">**3. Is (1 - upward probability) the downward probability?**</span>


## **Claude Response**

Strictly speaking, no.

Since a price's next movement has three possibilities — up, flat, and down — `1 - P(up)` = `P(flat) + P(down)`. The current model treats this as a binary score without distinguishing between them, so when the upward probability comes out at 40%, there's no way to tell whether it's "60% downward" or "high probability of staying flat."

There are three approaches for improvement.

First, switching to a 3-class model. Estimate the probability for each of up/flat/down and output them in a softmax format summing to 100%. Example: up 35%, flat 40%, down 25%.

Second, introducing thresholds. Define movements within ±0.1% of the current price as "flat," and classify only movements beyond that range as up or down.

Third, converting to expected return. Instead of probability, estimate the "expected price change over the next N minutes" to express both direction and magnitude simultaneously.

Shall I upgrade to a 3-class model?


## <span style="color: blue;">**4. Model Upgrade: 3-Class**</span>


## **Claude Work**

I'll rewrite the entire file as a 3-class model. The key changes are softmax-based 3-class probability, introduction of flat zone thresholds, and a 3-color probability trend chart. The upgrade to the 3-class model is complete. Here are the main changes.

**Model Structure Change**

Instead of the previous binary score (0–100%), a directional score is first calculated from the weighted sum of 6 factors, then combined with a volatility indicator to create 3 logits. The up logit increases when directionality is positive, the down logit when negative, and the flat logit strengthens when volatility is low and directionality is weak. Softmax is applied to these 3 logits to convert them into a probability distribution summing to 100%.

**Flat Detection Logic**

Flat strength = `(1 - volatility) × 0.7 + (1 - directionality strength) × 0.3`. This formula naturally increases the flat probability in zones where recent candle ranges are small and factors conflict, making directionality weak.

**UI Changes**

The gauge chart was replaced with a 3-class bar chart, and the probability trend chart was split into 3 lines (green/yellow/red). Directionality score, volatility level, and RSI are displayed as separate metric cards.
