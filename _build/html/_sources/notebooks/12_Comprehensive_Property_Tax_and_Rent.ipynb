## <span style="color: blue;">**Did the Comprehensive Property Tax Drive Up Rents?**</span>

<span style="color: blue;">An empirical analysis has found that increasing property holding taxes, including the Comprehensive Real Estate Tax (종부세, *Jongbuse*), leads to sharp rent increases. The finding suggests that landlords pass the holding tax burden on to tenants. According to the analysis, during the Roh Moo-hyun administration (2003–2008), when the Comprehensive Property Tax was introduced, monthly rents rose by over 20%. During the Moon Jae-in administration (2017–2022), when the tax was expanded and strengthened, rents also surged by more than 30%.</span>

<span style="color: blue;">The Comprehensive Property Tax was enacted on January 5, 2005, during the Roh Moo-hyun administration, and the tax burden was eased following the '9·1 Tax Reform Plan' on September 1, 2008, after the Lee Myung-bak administration took office. However, after the Moon Jae-in administration came to power, the tax rate was raised again through measures such as the 'Housing Market Stabilization Plan' on September 13, 2018, and the 'Housing Market Stabilization Measures' on December 16, 2019.</span>
 
<span style="color: blue;">On the 28th, according to academic circles, the research lab of Professor Lee Chang-moo at Hanyang University publicly released a study titled 'Real Estate Market Conditions and Policy Diagnosis in the Early Lee Jae-myung Administration' to the National Assembly on the 27th. To examine the pass-through effect of the Comprehensive Property Tax on rents, the study analyzed tax collection data from Seoul, Gyeonggi, and Incheon from 2005 to 2022, along with rent index and rent time-series data from 2006 to 2023.</span>
 
<span style="color: blue;">To isolate the relationship between Comprehensive Property Tax collections and rents, housing supply and demand as well as macroeconomic factors were treated as control variables. The rent level was measured separately using two indicators: the R114 rent index and a self-calculated actual transaction price index based on transaction prices registered with the Ministry of Land, Infrastructure and Transport (MOLIT). Both the R114 rent index and the actual transaction price index use January 2011 rent as the base value of 100, measuring relative levels. However, the analysis is limited in that the Comprehensive Property Tax collection statistics were only available through the end of 2022, so rent increases thereafter could not be analyzed.</span>

<span style="color: blue;">According to Professor Lee, the Comprehensive Property Tax collected in Seoul was 28.9 billion won in 2005, the year of introduction, but grew to 382.4 billion won in 2006 and nearly reached 1 trillion won at 903.7 billion won in 2007. In 2008, 610.7 billion won was collected. The decrease in 2008 collections reflects the impact of the global financial crisis and the phased implementation of the Lee Myung-bak administration's tax reduction policy (the 9·1 Tax Reform Plan). Total collections from 2005 to 2008 amounted to 1.9259 trillion won.</span>
 
<span style="color: blue;">Seoul's R114 rent index stood at 79.80 in January 2005. However, it rose to 80.70 by December 2006, then to 90.05 by December 2007, and further to 96.18 by the end of 2008. The cumulative index increase over this period was 20.5% (16.38 points).</span>
 
<span style="color: blue;">Over the same period, Comprehensive Property Tax collections in Gyeonggi Province also increased from 6.0 billion won in 2005 to 107.3 billion won in 2006, with 284.9 billion won collected in 2007 and 165.0 billion won in 2008. Total collections were 563.2 billion won. During this period, Gyeonggi Province's R114 rent index also rose from 82.91 in January 2005 to 97.73 at the end of 2008, an increase of 17.8% (14.82 points).</span>
 
<span style="color: blue;">In other words, rents in Seoul and Gyeonggi rose by approximately 20% following the introduction and strengthening of the Comprehensive Property Tax during the Roh Moo-hyun administration. Professor Lee explained, "Analysis and assessment of rent transaction prices reveals a pattern in which the Comprehensive Property Tax leads to sharp rent increases."</span>
 
<span style="color: blue;">A similar pattern appeared during the Moon Jae-in administration. The Moon administration, which took office in May 2017, implemented regulatory strengthening measures including Comprehensive Property Tax rate increases through the '9·13 Real Estate Measures,' '12·16 Real Estate Measures,' and other policies. This policy stance continued until the Yoon Suk-yeol administration finalized its 2022 Tax Reform Plan (July 21, 2022), which lowered the Comprehensive Property Tax rate for multi-home owners and raised the basic deduction threshold.</span>
 
<span style="color: blue;">In 2018, the year after the Moon Jae-in administration took office, the Comprehensive Property Tax assessed in Seoul was 275.4 billion won, but it increased to 619.3 billion won in 2019, and tax revenue continued to grow in 2020 (996.5 billion won) and 2021 (2.3741 trillion won). Total collections during this period amounted to 4.2655 trillion won.</span>
 
<span style="color: blue;">The R114 rent index was 103.74 in January 2018, only slightly above the base year of January 2011. However, after the Comprehensive Property Tax was strengthened, it rose to 114.36 by December 2021 and further to 123.46 in 2022, the last year of the Moon Jae-in administration. The cumulative rent index increase over this period was 19.0% (19.72 points).</span>

<span style="color: blue;">The sharp rent increase was also confirmed using the actual transaction price index calculated independently by Professor Lee's research lab based on MOLIT transaction data. The actual transaction rent index was 96.39 in January 2018, below the base year level of January 2011. However, by December 2022, this index had risen to 127.32. The cumulative increase was 32.08% (30.93 points). This means that someone paying 1 million won in monthly rent in 2018 would have had to pay 1.32 million won by the end of 2022.</span>
 
<span style="color: blue;">Professor Lee stated, "The introduction and strengthening of the Comprehensive Property Tax during the Roh Moo-hyun administration caused a rent surge of nearly 20%, and the re-strengthening of the tax during the Moon Jae-in administration caused a rent surge of 30–40% based on the actual transaction price index." He explained, "Since we used Comprehensive Property Tax collection statistics only through the end of 2022, we were unable to analyze rent trends in Seoul and other areas following the Yoon Suk-yeol administration's announcement of tax cuts." He added, "However, the recent rent increases are expected to be driven not only by the Comprehensive Property Tax factor but also by the effects of other regulations such as the government's lending restrictions." This study was presented as the keynote address at the 'Is Real Estate Policy Really Okay?' policy seminar held at the National Assembly Members' Office Building on the afternoon of the 27th.</span>

## **Claude's Assessment**

This study addresses an interesting topic, but has several important limitations in terms of methodology and interpretation.

**Strengths of the Study**

Using a long time series (2005–2022) and conducting parallel analysis with two rent indices (R114 and actual transaction-based) is commendable. The stated effort to control for housing supply/demand and macroeconomic variables also demonstrates basic methodological rigor.

**Key Problems**

First, the greatest issue is the **limitation in causal inference**. Simply observing that Comprehensive Property Tax collections and rents rose during the same period is insufficient to conclude that "the Comprehensive Property Tax caused rent increases." Both variables may move simultaneously due to a common cause: rising real estate prices. When home prices rise, both the tax base and the tax amount increase, while rents naturally rise as well. The possibility of **spurious correlation** is substantial.

Second, the **adequacy of control variables** is questionable. While the study claims to have "controlled for housing supply, demand, and macroeconomic factors," it is unclear exactly which variables were controlled and how. It is difficult to verify whether key factors directly affecting rents — such as the low interest rate environment, the trend of converting *jeonse* (lump-sum deposits) to monthly rent, and the implementation of the Three Rental Laws (2020) — were properly isolated.

Third, the study **overlooks the structural shift from *jeonse* to monthly rent**. The rent increases during the Moon Jae-in administration were heavily influenced by reduced *jeonse* supply and accelerated conversion to monthly rent following the Three Rental Laws. Failing to separate this effect from the Comprehensive Property Tax effect risks overestimating the tax's impact.

Fourth, there is **a problem with the choice of analysis periods**. The study spotlights only the Roh Moo-hyun period (2005–2008) and the Moon Jae-in period (2018–2022), but lacks symmetric analysis of how rents changed when the Comprehensive Property Tax was eased during the Lee Myung-bak and Park Geun-hye administrations. Looking at the graph, rents appear to have risen steadily even during periods of tax easing, which rather suggests that factors other than the Comprehensive Property Tax are the primary drivers of rent increases.

Fifth, as the researcher himself acknowledges, **the tax reduction effect of the Yoon Suk-yeol administration after 2022 could not be verified.** If rents continued to rise even after the Comprehensive Property Tax was reduced, the study's core argument would be weakened.

**Overall Assessment**

This study has value in presenting the hypothesis of "Comprehensive Property Tax → rent pass-through," but it tends to interpret correlation as causation, and lacks sufficient controls and comparisons for competing hypotheses (the *jeonse*-to-monthly-rent conversion, low interest rates, the Three Rental Laws, etc.). Given that this was a study presented at a policy seminar, it should be read with awareness that it may serve the purpose of supporting a particular policy direction.

## **ChatGPT's Assessment**

This analysis claims a **causal relationship** — "increased holding tax → sharp rent increase" — but from an academic standpoint, the methodology presented is closer to **describing a strong correlation, and is considerably weak as a causal inference design**. Let us organize the key issues from an econometric perspective.

---

### 1. What the Study Actually Did: Structure

The study has roughly the following structure:

* Explanatory variable: Comprehensive Property Tax collections in Seoul, Gyeonggi, and Incheon
* Dependent variable: Rent index

  * R114 rent index
  * Self-calculated index based on MOLIT actual transaction prices
* Period: 2005–2022
* Controls: Claims to have controlled for some supply, demand, and macroeconomic variables

And it draws its conclusion based on:

> The periods of sharp Comprehensive Property Tax increases (Roh Moo-hyun and Moon Jae-in administrations) coincide with periods of sharp rent increases → Tax pass-through occurred

This appears plausible at first glance, but **econometrically, this is one of the most dangerous analytical structures**.

---

### 2. The Biggest Problem: "Rents Did Not Rise Because Taxes Increased"

#### ❗ Core Identification Problem: **Endogeneity**

Comprehensive Property Tax collections appear to be a policy variable, but in reality:

$$\text{Tax collections} = f(\text{tax rate}, \text{assessed value}, \text{housing price level}, \text{number of homes})$$

In other words:

> **When home prices rise, tax collections automatically increase.**

Rents are also influenced by the same real estate market conditions as home prices.

Therefore, the observed relationship is likely due to:

> Rising real estate prices → Tax collections ↑
> Rising real estate prices → Rents ↑

There is a very high probability that this **common cause** is driving the pattern.

In this case, the estimation equation is effectively:

$$\text{Rent}_t = \beta \cdot \text{TaxCollections}_t + u_t$$

but the true structure is:

$$
\text{Rent}_t = \gamma \cdot \text{RealEstateConditions}_t + \epsilon_t
$$
$$
\text{TaxCollections}_t = \delta \cdot \text{RealEstateConditions}_t + \nu_t
$$

If $\text{RealEstateConditions}_t$ is omitted, $\beta$ will have **severe bias**.

👉 This is a textbook case of **omitted variable bias**.

---

### 3. Why "Including Control Variables" Does Not Resolve This

The study claims to have "controlled for supply, demand, and macroeconomic variables," but the following is important:

#### Key State Variables of the Real Estate Market Are Imperfectly Observed

* Expected price appreciation rates
* Investment demand sentiment
* *Jeonse*-to-monthly-rent conversion pressure
* Policy risk expectations

These are difficult to properly quantify.
The 2017–2022 period in particular saw overlapping structural changes such as lending regulations, *jeonse* loan policies, and the Three Rental Laws.

In short, **there are too many other shocks moving simultaneously with the tax.**

---

### 4. Using "Collection Amounts" Itself Is Highly Vulnerable

Tax pass-through studies typically examine:

* Changes in tax rates
* Changes in assessment criteria
* Policy changes that apply only to specific groups

However, this study uses **aggregate tax revenue (collection amounts)**.

This is a major problem:

> Collections = tax rate effect + tax base increase effect + taxpayer count increase effect

In other words, it is **impossible to separate whether the increase is due to policy or rising home prices**.

Academically, one should use **exogenous policy shocks** such as tax rate changes or assessment criteria changes.

---

### 5. Simple Time-Series Correlation Is Nearly Meaningless

This study essentially shows:

> Administration A: Taxes ↑ + Rents ↑
> Administration B: Taxes ↑ + Rents ↑

This is closer to a **regime comparison**.

But other things happening during the same periods:

#### Roh Moo-hyun Administration

* Rapid home price appreciation period
* Explosion of housing demand in the Capital Region
* Structural changes in the *jeonse* market

#### Moon Jae-in Administration

* Low interest rates + liquidity surge
* Three Rental Laws
* *Jeonse* price spike → conversion to monthly rent
* Lending regulations

In other words, "the period when the Comprehensive Property Tax rose" = "the period when the real estate market was undergoing upheaval."

Without separating these factors, the analysis captures a **period effect, not a policy effect**.

---

### 6. What a Genuine Tax Pass-Through Study Would Require

To be academically persuasive, the study would need at minimum:

#### ✔ (1) A Difference-in-Differences (DID) Framework

For example:

* Areas heavily affected by the Comprehensive Property Tax vs. areas with minimal impact
* Homes subject to the tax vs. homes not subject to the tax

$$
\text{Rent}_{it} = \alpha + \beta(\text{Taxed}_i \times \text{PostPolicy}_t) + \text{Fixed Effects} + \epsilon_{it}
$$

This would allow comparison between "homes that were taxed" and "homes that were not."

#### ✔ (2) Analysis Centered on Tax Rate Changes

Instead of collection amounts, one should use **policy design variables** such as:

* Magnitude of tax rate increases
* Changes in deduction thresholds
* Shifts in assessment bracket boundaries

#### ✔ (3) Explicit Modeling of Home Price and *Jeonse* Dynamics

Monthly rent is strongly linked to the *jeonse* market.
The key driver of the rent surge during the Moon Jae-in administration was:

> Decline in *jeonse* supply → conversion to monthly rent

Without removing this effect, the estimated tax effect is distorted.
