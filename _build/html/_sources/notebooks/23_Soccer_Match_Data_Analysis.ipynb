### 1. Overview

This document describes the structure and fields of the 2018 FIFA World Cup Final (France vs Croatia) match data provided in **StatsBomb Open Data** format. The data consists of two JSON files.

| File | Scale | Description |
|--------|------|------|
| `16-event_8658.json` | 2,978 events | Detailed records of all events occurring during the match (passes, shots, tackles, fouls, etc.) |
| `16-lineup8658.json` | 45 players | Player information, positions, and card records for both team squads (23 + 22 players) |

**Match result:** France 4 – 2 Croatia | **Formation:** France 4-4-2, Croatia 4-3-3 | **Coordinate system:** 120×80 yards

### 2. Event File (Event Data)

The event file is a JSON array that records all actions occurring during the match in chronological order. Each event consists of common fields and type-specific fields.

#### 2-1. Common Fields

Basic fields included in all events:

| Field | Type | Description |
|--------|------|------|
| `id` | UUID | Unique identifier for each event |
| `index` | Integer | Event sequence number (starting from 1) |
| `period` | Integer | Match half (1: first half, 2: second half) |
| `timestamp` | String | Time within the half (HH:MM:SS.sss) |
| `minute` / `second` | Integer | Minutes/seconds based on overall match time |
| `type` | Object | Event type (`id`, `name`) |
| `possession` | Integer | Possession sequence number |
| `possession_team` | Object | Team information for the possession |
| `play_pattern` | Object | Play pattern (open play, set piece, etc.) |
| `team` | Object | Team performing the event |
| `player` | Object | Player performing the event (`id`, `name`) |
| `position` | Object | Player's position at the time of the event |
| `location` | Array [x, y] | Pitch coordinates (based on 120×80) |
| `duration` | Float | Event duration (seconds) |
| `related_events` | Array | List of related event UUIDs |

#### 2-2. Event Type Distribution

Distribution of major event types among 32 total types:

| Event Type | Count | Description |
|-------------|------|------|
| Pass | 846 | Pass events (recipient, length, angle, height, outcome, etc.) |
| Ball Receipt* | 747 | Pass reception events |
| Carry | 617 | Ball carry (start/end coordinates) |
| Pressure | 254 | Pressing events against opponents |
| Ball Recovery | 88 | Ball recovery |
| Duel | 53 | 1v1 contests (aerial duels/tackles, etc.) |
| Clearance | 37 | Clearances |
| Shot | 23 | Shots (includes xG, outcome, freeze frame) |
| Goal Keeper | 32 | Goalkeeper actions (saves, punches, etc.) |
| Foul Committed | 28 | Fouls |
| Dribble | 28 | Dribble attempts |
| Interception | 17 | Interceptions |
| Other | 208 | Substitutions, injury stoppages, camera, tactical changes, etc. |

#### 2-3. Type-Specific Fields

▶ Pass

| Field | Description |
|------|------|
| `pass.recipient` (Object) | Receiving player |
| `pass.length` (Float) | Pass distance |
| `pass.angle` (Float) | Pass angle (radians) |
| `pass.height` (Object) | Height (Ground / Low / High) |
| `pass.outcome` (Object) | Outcome (absent = successful) |
| `pass.type` (Object) | Type (Corner, Free Kick, Goal Kick, etc.) |

▶ Shot

| Field | Description |
|------|------|
| `shot.statsbomb_xg` (Float) | Expected goals (xG) value |
| `shot.end_location` (Array [x,y,z]) | Shot destination coordinates (3D) |
| `shot.outcome` (Object) | Outcome (Goal, Saved, Off T, Blocked, Wayward) |
| `shot.technique` (Object) | Technique (Normal, Volley, Half Volley, etc.) |
| `shot.body_part` (Object) | Body part |
| `shot.freeze_frame` (Array) | Coordinates and positions of all players at the moment of the shot |

▶ Carry

| Field | Description |
|------|------|
| `carry.end_location` (Array [x, y]) | Ball carry end coordinates |

▶ Duel

| Field | Description |
|------|------|
| `duel.type` (Object) | Contest type (Aerial Lost/Won, Tackle) |
| `duel.outcome` (Object) | Outcome |

▶ Goal Keeper

| Field | Description |
|------|------|
| `goalkeeper.type` (Object) | Action type (Shot Faced, Save, Punch, etc.) |
| `goalkeeper.position` (Object) | Stance (Set, Moving) |
| `goalkeeper.end_location` (Array) | End coordinates |

▶ Substitution

| Field | Description |
|------|------|
| `substitution.outcome` (Object) | Reason (Tactical, Injury) |
| `substitution.replacement` (Object) | Replacement player |

#### 2-4. Play Pattern

Each event records the type of play initiation:

`Regular Play` · `From Corner` · `From Counter` · `From Free Kick` · `From Goal Kick` · `From Keeper` · `From Kick Off` · `From Throw In` · `Other`

#### 2-5. Coordinate System

StatsBomb uses a **120 × 80 yards** coordinate system. The attacking direction is always left to right (x: 0→120), and the y-axis goes from 0 (top) to 80 (bottom). The `end_location` for shots is three-dimensional (x, y, z), where z represents height.

---

### 3. Lineup File (Lineup Data)

The lineup file contains complete squad information for both teams. It includes data for a total of 45 players: 23 from France and 22 from Croatia.

#### 3-1. Field Structure

| Field | Description |
|--------|------|
| `team_id` / `team_name` | Team identification |
| `player_id` / `player_name` | Player identification |
| `player_nickname` | Player nickname (e.g., Kylian Mbappé) |
| `jersey_number` | Jersey number |
| `country` | Player nationality |
| `cards` | List of yellow/red cards |
| `positions` | Position change history (with time and reason) |

### 3-2. Position Change History (positions)

The `positions` array records player position changes during the match in chronological order. Each entry includes `position` (position name), `from`/`to` (time), `from_period`/`to_period` (half), and `start_reason`/`end_reason` (reason for change: Starting XI, Substitution, Tactical Shift, Final Whistle, etc.).

**Example:** Kylian Mbappé started as Right Midfield, then changed to Right Center Forward via Tactical Shift at the 81st minute, maintaining that position until the end of the match.

---

### 4. Potential Use Cases

- **Match flow analysis:** Visualize match flow using `timestamp` and `possession` sequences
- **Pass network:** Construct pass networks from passer→recipient relationships in Pass events
- **xG model validation:** Expected goals analysis using `statsbomb_xg` and `freeze_frame` from Shot events
- **Defensive analysis:** Measure defensive intensity using Pressure, Duel, Interception, and Ball Recovery data
- **Heatmaps / Visualization:** Spatial analysis of shots, passes, pressing, etc. using `location` coordinates
- **Player evaluation:** Individual performance analysis combining Lineup position change history + event data

---

### 5. Notes

- **Data source:** StatsBomb Open Data (https://github.com/statsbomb/open-data)
- The `*` in `Ball Receipt*` indicates that the event may have been an incomplete reception (miscontrol).
- `under_pressure` (Boolean) is additionally flagged for events performed under opponent pressure.
- `related_events` allows identification of connections between events (e.g., pass ↔ reception, shot ↔ save).
- **xG (Expected Goals):** France 1.10, Croatia 1.48 — Croatia led in xG but the actual result was reversed.

## **Claude: Analysis**

An individual performance analysis dashboard was created for all 27 players who appeared in the match. Clicking each player card expands the detailed analysis.

> 🔗 **2018 FIFA World Cup Final Player Evaluation Dashboard** (https://claude.ai/public/artifacts/b878065b-1feb-4837-8014-bd686662baec)

![Worldcup](https://raw.githubusercontent.com/pilsunchoi/images4/main/23-3.png)

**Radar Chart** — Visualizes percentile rankings relative to starting players on a Per 90-minute basis across six axes: Attack, Passing, Carrying, Pressing, Defense, and Volume. Pogba shows extreme values on both the attacking and defensive fronts, while Brozović shows extreme values in passing and volume.

**Position Change History** — Displays the `positions` array from the Lineup data as a timeline. For example, Mbappé switched from Right Midfield → Right Center Forward at the 81st minute, and Griezmann also changed positions at the same time, allowing tracking of Didier Deschamps' second-half tactical adjustments.

**Key Findings:**
- **Brozović** — 271 touches and 99 passes (87 successful), the most in the match. The absolute hub of Croatia's build-up play.
- **Mbappé** — 6 progressive dribbles and 245 yards of carry distance, the core of France's counter-attacks. Scored 1 goal from an xG of 0.06 (exceeding expected finishing).
- **Griezmann** — 31 pressures, the most for France. Despite being an attacker, his forward defensive contribution was overwhelming.
- **Kanté** — 19 pressures in 54 minutes played, plus 1 yellow card. This suggests his substitution was motivated by fitness management and card avoidance.
