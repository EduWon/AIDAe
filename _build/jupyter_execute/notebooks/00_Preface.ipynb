# **Preface** 


Whether it's a school assignment or a work task, we now live in a world where people instinctively ask AI first before starting anything. Today's AI delivers excellent answers quickly and reliably. Data analysis may well be the ultimate showcase for this capability. Just a year or two ago, AI was limited to answering questions about analysis ideas, coding, and statistical knowledge. Now, you simply feed it data, and AI autonomously produces expert-level reports and even interactive dashboards. There are no restrictions on the type of data either. Beyond structured data in spreadsheet format, AI can now handle text, images, and video — the horizon of data analysis has expanded beyond comparison. This book aims to demonstrate AI's current data analysis capabilities through illustrative examples. We hope readers will be able to gauge the current level of AI-powered data analysis and find inspiration for applying it to their own tasks.

We asked generative AI to perform the following data analysis tasks as illustrative examples.

* Exam score analysis for a university class
* Presidential approval rating changes by age and region
* 2025 KBO regular season team performance analysis
* PPP theory graph analysis and update
* Collection and analysis of major countries' GDP growth rates before and after COVID-19
* Building a stock price prediction system using order book and candlestick charts
* Comment analysis on a news article about the rush toward medical schools
* Detection analysis of daily apartment listing changes in Seoul
* Airline passenger satisfaction survey analysis
* Regression analysis of Boston housing prices
* Evaluating media reports on the empirical relationship between comprehensive property tax and rent
* Empirical analysis methodology for the decline of hypermarkets
* Evaluating media reports on the empirical relationship between newborn special loans and birth rate
* Logistic regression analysis of bank telemarketing data
* Tree model-based Titanic passenger survival classification
* Deep learning-based handwritten digit classification
* Panel data analysis of wage determination models
* 2018 World Cup final match data analysis
* Estimating shock propagation patterns from the Iran bombing
* Global cosmetics trade trend analysis
* Texas special education status and open data survey

We provided Claude, ChatGPT, and Gemini with various types of data related to the above topics and asked them to analyze and explain the results on their own. The key point of this book is that we did not ask AI about statistical knowledge or programming code — we had AI perform the analysis "directly."

The data we provided (uploaded) to AI included structured data such as Excel, CSV, and JSON files, but also frequently included unstructured data such as text, images, and video. The power of multimodal AI models truly shines in unstructured data analysis.

In some cases, we did not provide any data to analyze and instead asked AI to find and analyze the data on its own. Examples include the collection and analysis of major countries' GDP growth rates before and after COVID-19, global cosmetics trade trend analysis, and the Texas special education status and open data survey.

Our analysis requests (prompts) were also diverse. In some cases, we provided very specific directions for the analysis, while in others, we simply asked "analyze this." We also asked AI to evaluate empirical analyses that others had already conducted. The evaluation of media reports on the relationship between comprehensive property tax and rent, and between newborn special loans and birth rate, fall into this category.

You can see the remarkable analytical capabilities AI demonstrates across these diverse scenarios in the 26 chapters of this book. Even with vague requests, AI produces expert-level reports in just one or two minutes. It is also highly impressive how AI presents analysis and research results intuitively using interactive dashboards.

While AI possesses outstanding capabilities, we also witnessed it making mistakes and sometimes presenting incorrect results. One must be wary of so-called hallucinations, where AI provides plausible-sounding explanations for things that are not true. Furthermore, AI generates responses that reflect the user's questioning style and context. Ultimately, what matters is not only the ability to use AI effectively, but also the researcher's competence to critically examine and filter the results.

At the end of each chapter, we have included the Python code used for the analysis. AI sometimes provides the code along with the analysis results, but when it does not, you can simply request it separately. AI's Python execution environment (sandbox) is restricted to pre-installed libraries such as `pandas`, `numpy`, `matplotlib`, and `scikit-learn` for security and stability. Therefore, for packages that are not pre-installed (e.g., `linearmodels` in Chapter 20), AI cannot install them and can only provide us with the code. In such cases, the user must run the code in their local environment and feed the results back to AI.

The analyses in this book were conducted using (as of the time of writing) Claude Opus 4.6, GPT-5.2, and Gemini 3 Pro. They are the co-workers of this book.
