## <span style="color: blue;">**2. Ridge Regression**</span>

## **Gemini Analysis**

### 1. Understanding Ridge Regression

Ridge regression is a regularization technique that constrains the magnitude of coefficients to prevent overfitting in linear regression models. Standard Ordinary Least Squares (OLS) can produce excessively large coefficient values in the process of minimizing the residual sum of squares; Ridge regression controls this by adding an $\ell_2$ penalty (the sum of squared coefficients) to the objective function.

The objective function of Ridge regression is:

$$\min_{\beta} \left\{ \sum_{i=1}^{n} (y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij})^2 + \lambda \sum_{j=1}^{p} \beta_j^2 \right\}$$

The first term in this expression is the residual sum of squares (RSS), representing the error between the model's predictions and actual values. The second term is the penalty term, which multiplies the sum of squared coefficients by the hyperparameter $\lambda$ (or alpha).

$\lambda$ controls the strength of regularization; larger values shrink the coefficient estimates closer to 0, making the model simpler. As a result, Ridge regression improves the model's generalization performance by finding a balance between reducing errors and keeping coefficient magnitudes small.

In this process, some **regularization bias** may be introduced as a trade-off for reducing model variance, but the overall effect is to improve prediction accuracy (MSE). This is particularly useful when multicollinearity (high correlations among variables) is present.

### 2. Estimation Procedure

The analysis proceeds in four steps:

1.  **Data standardization**: Since variables have different units, penalties would be applied unfairly; therefore, all independent variables are transformed to have mean 0 and standard deviation 1.
2.  **Hyperparameter tuning**: Cross-validation is performed to find the $\lambda$ (or alpha) value that determines model complexity. In this analysis, 5-fold cross-validation was used to select the optimal $\lambda \approx 2.477$ that yields the lowest error.
3.  **Model training**: Using the selected $\lambda$, regression coefficients are estimated on the full dataset.
4.  **Statistical measures**: Standard errors, t-values, and p-values are calculated to evaluate the reliability of the estimated coefficients.

## <span style="color: blue;">**3. Constraining Coefficient Magnitudes to Prevent Overfitting?**</span>

## **Gemini Explanation**

Constraining coefficient magnitudes to prevent overfitting means suppressing the **model's flexibility** so that it does not overreact to **noise** in the training data.

### 1. What Large Coefficients Mean: High Sensitivity

In a regression model, the coefficient $\beta$ represents how much the dependent variable $y$ changes when the independent variable $x$ changes by 1 unit.

* **What if coefficients are very large?**: Even a tiny change in $x$ causes $y$ to change drastically. If certain observations in the training data contain noise from measurement errors or chance, the model inflates the coefficients to abnormal levels to fit that noise.
* **Result**: The model ends up chasing the 'whims' of individual data points rather than the general pattern, causing prediction accuracy to drop sharply when new data (test data) is introduced.

### 2. Geometric Intuition of Coefficient Constraints: From Curves to Straight Lines

When drawing a function that passes through data points:

* **Without constraints**: The model draws a complex, sinuous curve to perfectly pass through every point. The degree of this sinuosity is directly related to the magnitude of the coefficients.
* **With coefficient constraints**: When a penalty is applied to the sum of squared coefficients ($\ell_2$ norm) as in Ridge regression, the model strives to keep coefficients as small as possible. This has the effect of smoothing out the sinuous curve, making it as gentle or close to a straight line as possible.
* **Meaning**: It is equivalent to declaring, "Ignore the minor bumps (noise) in the data and follow only the overall trend (signal)."

### 3. Regularization and the Bias-Variance Tradeoff

Constraining coefficients is statistically a strategy of **reducing variance while slightly increasing bias**.

1.  **Variance reduction**: The model becomes less sensitive to small changes in the data, producing similar estimates regardless of which training data is used. In other words, the model becomes stable.
2.  **Bias increase**: Because coefficients are forcibly reduced, the model cannot perfectly match the training data targets. This is called **regularization bias**.

Ultimately, the reason for constraining coefficients is: **"To dramatically improve prediction stability in practice (variance reduction), even at the cost of a slight accuracy loss (bias increase)."**

### 4. Mathematical Expression

The meaning of coefficient constraint becomes clearer when examining the Ridge regression objective function:

$$Cost = \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 + \lambda \sum_{j=1}^{p} \beta_j^2$$

The model must minimize the sum of these two terms. As $\lambda$ increases, the model focuses on keeping coefficients ($\beta$) small, even if it means tolerating slightly more error.

---

**In summary, coefficient constraint is a mechanism that ensures the model learns robust general rules without being swayed by trivial features of the data.**

## <span style="color: blue;">**4. Hyperparameter Tuning via Cross-Validation**</span>

## **Gemini Explanation**

### Hyperparameter Tuning: 5-Fold Cross-Validation Procedure

One of the most important tasks in Ridge regression is determining the value of the hyperparameter $\lambda$ (or alpha), which controls the strength of regularization. If $\lambda$ is too small, overfitting occurs; if too large, the model becomes too simple, resulting in underfitting. The step-by-step procedure of 5-fold cross-validation used to address this is as follows.

---

### Step 1: Set the Range of Candidate $\lambda$ Values
First, define a set of candidate $\lambda$ values to explore. Typically, a wide range on a log scale is specified, such as `0.001`, `0.01`, `0.1`, `1`, `10`, `100`, to identify which order of magnitude yields good model performance.

### Step 2: Split the Dataset (Create 5 Folds)
Divide the entire training data into 5 groups (folds) of roughly equal size. These are called $F_1, F_2, F_3, F_4, F_5$.



### Step 3: Iterative Training and Validation (Loop Execution)
For each candidate $\lambda$ value, repeat the following process 5 times:

* **Iteration 1**: Train the model on $F_2, F_3, F_4, F_5$ and evaluate prediction performance (e.g., MSE) on the held-out $F_1$.
* **Iteration 2**: Train the model on $F_1, F_3, F_4, F_5$ and evaluate performance on the held-out $F_2$.
* **...**
* **Iteration 5**: Train the model on $F_1, F_2, F_3, F_4$ and evaluate performance on the held-out $F_5$.

This process yields a total of 5 performance metric results for each $\lambda$ value.

### Step 4: Calculate Average Performance
Compute the arithmetic mean of the 5 performance metrics obtained for each candidate $\lambda$.

$$\text{Average Error}(\lambda) = \frac{1}{5} \sum_{i=1}^{5} \text{Error}_i$$

This average represents the "typical performance" the model would exhibit on new data when using a specific $\lambda$. By splitting the data for validation, we prevent the model from being optimized to only a specific segment of the data.

### Step 5: Determine the Optimal $\lambda$ and Build the Final Model
Among all candidate $\lambda$ values, select the one with the lowest average error as the **optimal hyperparameter**. Finally, retrain the model using this optimal $\lambda$ on the entire training data (all 5 folds included) to estimate the final regression coefficients.

---

**In summary, 5-fold cross-validation is a process of rotating through 5 data splits — repeating training and validation — to find the optimal 'regularization strength ($\lambda$)' that delivers stable performance regardless of what data comes in.**

## <span style="color: blue;">**5. LASSO**</span>

## **Gemini Explanation**

### 1. Understanding Lasso Regression

Lasso (Least Absolute Shrinkage and Selection Operator) regression, like Ridge regression, is a linear model that uses regularization to prevent overfitting. However, unlike Ridge regression which uses the sum of squared coefficients ($\ell_2$ penalty), Lasso uses the **$\ell_1$ penalty** — the sum of the absolute values of the coefficients.

The objective function of Lasso regression is:

$$\min_{\beta} \left\{ \sum_{i=1}^{n} (y_i - \beta_0 - \sum_{j=1}^{p} \beta_j x_{ij})^2 + \lambda \sum_{j=1}^{p} |\beta_j| \right\}$$

The most distinctive feature of Lasso regression is its **feature selection** capability. When $\lambda$ is sufficiently large, it shrinks the coefficients of less influential variables to exactly 0, effectively removing them from the model. This makes the model simpler and more interpretable, achieving a substantial reduction in model variance at the cost of accepting some **regularization bias** from unimportant variables.

### 2. Estimation Procedure

The estimation of the Lasso model proceeds through the following steps:

1.  **Variable standardization**: Since the $\ell_1$ penalty also depends on the absolute magnitude of coefficients, all independent variables are standardized to have mean 0 and standard deviation 1 to ensure fair penalty application.
2.  **Hyperparameter ($\lambda$) tuning**: 5-fold cross-validation is performed to find the optimal regularization strength. The average prediction error is calculated across various $\lambda$ candidates, and the $\lambda$ with the smallest error is selected.
3.  **Variable selection and coefficient estimation**: The model is trained under the optimal $\lambda$. Variables with low contribution to housing price determination have their coefficients driven to 0 and are dropped, leaving only the important variables in the model.
4.  **Statistical significance evaluation**: For the surviving variables (those with non-zero coefficients), standard errors, t-values, and p-values are calculated to test the statistical significance of each factor.

## **Appendix: Python Code**

```python
import pandas as pd

url = "https://raw.githubusercontent.com/selva86/datasets/master/BostonHousing.csv"
df = pd.read_csv(url)
df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Ridge, Lasso, LassoCV
from sklearn.metrics import mean_squared_error, r2_score
from scipy import stats

# 1. Data preprocessing
X = df.drop('medv', axis=1)
y = df['medv']

# Standardize independent variables (essential for Ridge/Lasso)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
n, p = X_scaled.shape
```

```python
# 2. Ridge regression analysis
# Hyperparameter (lambda) tuning: 5-fold cross-validation
alphas = np.logspace(-3, 3, 100)
ridge_grid = GridSearchCV(Ridge(), param_grid={'alpha': alphas}, cv=5)
ridge_grid.fit(X_scaled, y)
best_alpha_ridge = ridge_grid.best_params_['alpha']

# Train optimal model
ridge_model = ridge_grid.best_estimator_
ridge_coefs = ridge_model.coef_

# Compute Ridge statistics (SE, t-value, p-value)
XtX = X_scaled.T @ X_scaled
I = np.eye(p)
inv_matrix_ridge = np.linalg.inv(XtX + best_alpha_ridge * I)
y_pred_ridge = ridge_model.predict(X_scaled)
rss_ridge = np.sum((y - y_pred_ridge)**2)

# Compute effective degrees of freedom
H_lambda = X_scaled @ inv_matrix_ridge @ X_scaled.T
df_eff = np.trace(H_lambda)
sigma_sq_ridge = rss_ridge / (n - df_eff)
cov_beta_ridge = sigma_sq_ridge * (inv_matrix_ridge @ XtX @ inv_matrix_ridge)
se_ridge = np.sqrt(np.diag(cov_beta_ridge))
t_stats_ridge = ridge_coefs / se_ridge
p_values_ridge = [2 * (1 - stats.t.cdf(np.abs(t), n - df_eff)) for t in t_stats_ridge]

# Ridge results DataFrame
ridge_results = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': ridge_coefs,
    'Std.Error': se_ridge,
    't-value': t_stats_ridge,
    'p-value': p_values_ridge
})

# Display results
print("\n--- Ridge Results ---\n")
print(ridge_results)
```

```python
# 3. Lasso regression analysis

# Hyperparameter (lambda) tuning: using LassoCV
lasso_cv = LassoCV(alphas=np.logspace(-4, 1, 100), cv=5, random_state=42)
lasso_cv.fit(X_scaled, y)
best_alpha_lasso = lasso_cv.alpha_

# Train optimal model
lasso_model = Lasso(alpha=best_alpha_lasso)
lasso_model.fit(X_scaled, y)
lasso_coefs = lasso_model.coef_

# Compute Lasso statistics (Post-Lasso approximation for non-zero variables)
non_zero_idx = np.where(lasso_coefs != 0)[0]
X_selected = X_scaled[:, non_zero_idx]
n, p_sel = X_selected.shape
X_sel_inc = np.column_stack([np.ones(n), X_selected]) # Include intercept
y_pred_lasso = lasso_model.predict(X_scaled)
df_resid_lasso = n - p_sel - 1
sigma_sq_lasso = np.sum((y - y_pred_lasso)**2) / df_resid_lasso
inv_xtx_lasso = np.linalg.inv(X_sel_inc.T @ X_sel_inc)
se_features_lasso = np.sqrt(np.diag(inv_xtx_lasso) * sigma_sq_lasso)[1:]

# Organize Lasso results
lasso_results_list = []
se_ptr = 0
for i, col in enumerate(X.columns):
    coef = lasso_coefs[i]
    if coef != 0:
        se = se_features_lasso[se_ptr]
        t_val = coef / se
        p_val = 2 * (1 - stats.t.cdf(np.abs(t_val), df_resid_lasso))
        se_ptr += 1
    else:
        se, t_val, p_val = np.nan, np.nan, np.nan
    lasso_results_list.append({'Feature': col, 'Coefficient': coef, 'Std.Error': se, 't-value': t_val, 'p-value': p_val})
lasso_results = pd.DataFrame(lasso_results_list)

# Display results
print("\n--- Lasso Results ---\n")
print(lasso_results)
```
