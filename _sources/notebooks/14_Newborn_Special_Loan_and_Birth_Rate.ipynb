Below is a detailed explanation of each of the three items you listed — why and how they are needed to examine **causally** whether the Newborn Special Loan raised birth rates. (I will use terminology as it appears in practice and academic papers.)

---

### 0. First: Why the Current Chart Is Not 'Evidence'

The year-by-year observation that "birth rates rose among the upper deciles" is a **pattern (correlation)**. However, other changes that occurred simultaneously with the Newborn Special Loan in 2024–2025 (interest rates, economic conditions, marriage/divorce trends, housing prices, other policies, etc.) may have disproportionately affected the upper deciles. In that case, attributing the birth rate increase to the loan policy would be a misjudgment.

What is needed, then, is a structure that constructs "what would have happened without the policy (**counterfactual**)," and **DID (Difference-in-Differences)** is one of the most standard methods for doing so.

---

### 1. Income Decile × Year DID — "Comparing Pre-Post Changes" by Income Group

#### 1.1 The Skeleton of Basic DID

The simplest DID looks like this:

* Groups: Treated vs. Control
* Time: Pre-policy vs. Post-policy
* Core:

$$
DID = (Y_{T,post}-Y_{T,pre}) - (Y_{C,post}-Y_{C,pre})
$$

where $Y$ is the outcome variable such as the total fertility rate (TFR).

What this says is:
"The treated group increased more after the policy, and this increase is not simply a time trend but an additional change occurring only in the treated group."

---

#### 1.2 What Does It Mean to Use "Income Decile × Year" as DID?

Using the 10 income deciles directly gives 10 'groups.' This means the analysis is not "one group vs. one group" but rather a **multi-group DID (panel / repeated cross-section)** format.

The most common regression specification is:

$$
Y_{d,t}=\alpha + \sum_{d\neq d_0}\gamma_d \cdot 1[d] + \sum_{t\neq t_0}\delta_t \cdot 1[t] + \beta \cdot (Post_t \times Treat_d) + \varepsilon_{d,t}
$$

* $d$: Income decile (1–10)
* $t$: Year (2006–2025)
* $1[d]$: Decile fixed effects (removes inherent differences in baseline fertility levels across deciles)
* $1[t]$: Year fixed effects (removes year-specific shocks common to all deciles: economic conditions, COVID, interest rates, etc.)
* $Post_t$: Equals 1 if year ≥ 2024
* $Treat_d$: Equals 1 for deciles expected to experience the policy effect (e.g., 8th–10th deciles for the top 30%)

$\beta$ is the DID estimate.
It captures "how much more the birth rate changed in the treated (upper) deciles relative to the comparison (lower/middle) deciles after the policy."

---

#### 1.3 Why Can't We Simply Compare 2023 → 2025 and Need DID?

Even if birth rates rose among the upper deciles in 2024–2025, during the same period:

* Upper deciles may have married more
* Upper deciles may have moved/purchased homes more
* The share of adults in their 30s may have coincidentally increased in upper deciles
* High-income job/wage growth may have concentrated in upper deciles

DID, at minimum, removes "year-common shocks" and "decile-fixed differences" and examines whether any additional change remains.

---

#### 1.4 Assumption Required for DID: Parallel Trends

Core assumption: Even without the policy, the outcomes for the treated and control groups would have **followed similar trends in the pre-policy period**.

A typical method for verifying this is the **event study**:

$$
Y_{d,t}=\alpha + \gamma_d + \delta_t + \sum_{k\neq -1}\beta_k \cdot (1[t - t_0 = k]\times Treat_d)+\varepsilon_{d,t}
$$

* $t_0$: Policy start year (2024)
* $k$: Year relative to policy start (−3, −2, −1, 0, 1, 2, …)

The key question here is **whether the $\beta_k$ values for the pre-policy period ($k<0$) are close to 0**.
If the upper deciles were already on an upward trend before the policy, the post-policy increase cannot be attributed to the policy.

---

#### 1.5 Practical Design Considerations (Important)

* **Whether to set Post = 1 at 2024** — since 2024 is the early implementation phase with monthly variation, one might consider setting post = 1 from 2025 (based on "actual exposure")
* Fertility has a **time lag** (conception to birth). A 2024 policy may be more fully reflected in 2025 births
* If the unit is individuals: various models are possible — birth occurrence (0/1), birth count, first-birth hazard, etc.
* If the unit is decile × year aggregates: the sample becomes small (10 × 20 = 200), potentially weakening standard errors and estimation stability

---

### 2. Policy-Eligible vs. Ineligible Group Comparison — Defining the "Actual Treatment" More Precisely

The key point here: "top 30% income" does **not map 1:1 to policy eligibility**.

The Newborn Special Loan depends not only on income but also (typically) on qualification criteria such as loan status, housing conditions, credit, and household composition. Therefore, the "policy-exposed population" must be defined more precisely.

---

#### 2.1 Why Is an 'Eligibility'-Based Comparison Important?

To examine the policy effect, the treated group should be "those who actually used or at least could have used the policy."

For example:

* Upper-income individuals who already own sufficient housing may have low demand for the loan, resulting in a small effect
* Middle-income individuals who are non-homeowners, newlyweds, and have clear childbearing plans may have high demand
* Credit/debt constraints may mean that "the program exists but is difficult to use" for low-income groups

Therefore, comparing **(i) eligible groups (meeting qualification requirements)** vs. **(ii) ineligible groups (outside the requirements from the start)** offers much stronger identification.

---

#### 2.2 Cleanest Design: RDD/DID-RDD Using the Income Cutoff

The "relaxation of the annual income requirement from 130 million to 200 million won (December 2024)" you mentioned is a particularly useful signal. Such cutoffs enable the following designs:

* Examining people near the 130 million won threshold (e.g., 120–140 million won), where eligibility sharply diverges before and after the policy
* The same applies near the 200 million won threshold

Possible approaches:

* **RDD (Regression Discontinuity Design)**: Comparison just above/below the cutoff
* **DID-RDD**: Comparing near the cutoff before and after the policy (removing time-specific shocks)

If feasible, this is far stronger causally than a crude classification like "top 30%."

---

#### 2.3 Second-Best Approach When Data Is Limited: Predicted Eligibility Probability (Propensity/Exposure)

When eligibility criteria are complex and individual conditions cannot be fully obtained, one can estimate "eligibility likelihood" using pre-policy characteristics and use exposure as a continuous variable.

* For example, using non-homeowner status, *jeonse* residence, location in high-price housing areas, income, debt, etc.
* Construct a "policy exposure index"

$$
Y_{i,t} = \alpha + \gamma_i + \delta_t + \beta(Post_t \times Exposure_i) + \varepsilon_{i,t}
$$

Here, $\beta$ examines "whether those with 1 unit greater exposure had a larger post-policy increase in fertility."

---

#### 2.4 Common Pitfalls in "Eligible vs. Ineligible" Comparisons

* **Endogenous selection**: Actual loan users may have had stronger fertility intentions from the start
  → Comparing "actual users" and "non-users" is therefore risky
  → Better to use "eligibility" or "exposure" as an instrument-like variable
* **Simultaneous policy changes**: If other fertility/housing policies also changed during the same period, confounding may occur
  → Need to verify whether the comparison group was equally exposed to such policies

---

### 3. Controlling for Marriage Rate, Age Structure, and Economic Conditions — Removing "Other Reasons for Upper-Income-Only Increases"

If year fixed effects are already included in the DID, "economic shocks common to all groups" are largely removed. But the problem is:

* Shocks may operate **differently across income deciles**
* Demographic composition changes (age, marriage) may move differently across deciles

If these are not controlled, they bleed into $\beta$ (omitted variable bias).

---

#### 3.1 Controlling for Marriage Rate (or Marital Status)

Fertility is strongly linked to marriage (especially for first births). If marriages increased among upper deciles in 2024–2025, the fertility increase could be due to marriage changes rather than the policy.

Possible variables:

* Decile × year marriage rate
* Birth rates by marital status (unmarried/married)
* First marriage rate, average age at first marriage
* Marriage-to-birth lag consideration (1–2 years after marriage)

How to include in regression (aggregate-level example):

$$
Y_{d,t}=\alpha+\gamma_d+\delta_t+\beta(Post_t\times Treat_d) + \theta \cdot MarriageRate_{d,t} + \varepsilon_{d,t}
$$

Caution: The marriage rate may itself be an outcome of the policy (if the policy increased marriages which in turn increased births). In that case, controlling for marriage rate leaves only the "direct effect" rather than the "total effect."
In practice, both are typically presented:

* (a) Total effect estimate without controlling for marriage
* (b) Mechanism decomposition with marriage control

---

#### 3.2 Controlling for Age Structure (Especially Distribution of Women Aged 20–39)

Birth rates are driven not only by 'behavior' but also by 'composition.'

Example: If the share of women aged 30–34 increases in the upper deciles, TFR can rise even with identical fertility behavior. Conversely, a decline in women in their 20s can lower it.

Possible controls:

* Share of women of childbearing age (15–49) in the 25–29, 30–34, 35–39 brackets
* Mean age at childbirth
* Age standardization: fixing age structure to compare "behavior only"

The strongest approach is to analyze using **age-specific fertility rates (ASFR)**:

* $Y_{a,d,t}$: Fertility rate for age group $a$
* This substantially reduces the age structure problem (because comparison is within age groups)

$$
Y_{a,d,t} = \alpha + \gamma_{a,d} + \delta_t + \beta(Post_t \times Treat_d) + \varepsilon_{a,d,t}
$$

(where $\gamma_{a,d}$ represents age × decile fixed effects)

---

#### 3.3 Controlling for Economic Conditions: "Where Year Fixed Effects Are Not Enough"

Year fixed effects remove economy-wide shocks common to all groups. But the problem is that upper and lower deciles may have different sensitivities to economic fluctuations.

Example: Upper deciles may be more sensitive to asset prices (stocks/real estate), while lower deciles may be more sensitive to unemployment/wages.

At the decile × year level, the following can be added:

* Decile-specific unemployment/employment rates
* Decile-specific real income growth (wages)
* Asset price indicators (housing prices, interest rates) interacted with deciles
* Adding regional fixed effects strengthens identification further (region × year shocks)

For example:

$$
Y_{d,t}=\alpha+\gamma_d+\delta_t+\beta(Post_t\times Treat_d)+\phi X_{d,t}+ \varepsilon_{d,t}
$$

or

$$
Y_{r,d,t}=\alpha+\gamma_{r,d}+\delta_t+\beta(Post_t\times Treat_d)+\phi X_{r,t}+ \varepsilon_{r,d,t}
$$

(Adding region $r$ improves identification)

---

#### 3.4 The Purpose of Controls Is "Identification," Not "Explanation"

The reason for including controls is not to "make the model look plausible," but to prevent the treatment effect $\beta$ from being spuriously inflated or deflated by changes in other factors (marriage/age/economic conditions).

---

### 4. Recommended Analysis Flow Combining All Three (Practical Approach)

1. **Baseline charts/summary statistics**: TFR trends by decile
2. **Event study** to check parallel trends (are pre-policy coefficients ≈ 0?)
3. **Basic DID**: Post × Treat (Treat = top 30%, etc.)
4. **Alternative definitions**: Redefine Treat as "eligible" (using cutoff if possible)
5. **Add controls**: Add marriage/age/economic variables and check stability of $\beta$
6. **Sensitivity/placebo**: Does the effect appear with a fake policy year (e.g., 2022)? If so, this is a warning sign
7. **Heterogeneity**: First vs. second child, age group (30–34), region (high housing price areas), etc.

#### 3.1 Person-Month Linear Probability Model (LPM)

$$
Birth_{i,t}=\alpha_i+\delta_t+\beta\,(Post^{A}_t \times Eligible^{A}_{h(i),t}) + \Gamma'X_{i,t} + \varepsilon_{i,t}
$$

- $\alpha_i$: Individual fixed effects (removes time-invariant birth propensity)
- $\delta_t$: Month fixed effects (absorbs nationwide common shocks: interest rates, COVID, policy bundles, etc.)
- $X_{i,t}$: Time-varying controls (marital-household status, employment eligibility, age group, etc.)

The $\beta$ in this model represents how much more the birth probability increased for the eligible group relative to the non-eligible group after the policy introduction.

---

#### 3.2 Event Study (Parallel Trends Verification + Lag Reflection)

Since fertility involves substantial lags, a single Post indicator is insufficient; relative-time coefficients are needed.

$$
Birth_{i,t}=\alpha_i+\delta_t+\sum_{k\neq -1}\beta_k\,(1[\mathrm{event\_time}(t)=k]\times Eligible^{A}_{h(i),t})+\Gamma'X_{i,t}+\varepsilon_{i,t}
$$

- event_time is the relative month with January 2024 as 0 (e.g., −24 to +18)
- If $\beta_k$ values are near 0 in the pre-policy period ($k<0$), parallel trends are plausible
- In the post-policy period ($k>0$), one can verify when effects begin to appear
- Realistically, fertility has a minimum 9–12 month lag, so the key is whether strong responses appear in the $k=9$ to $k=18$ window

---

#### 3.3 Standard Errors

- Minimum: Individual-level clustering (panel correlation)
- If possible: Household/family-level clustering (spouse-sibling correlation) should also be considered
  In practice, individual-level clustering + absorbing aggregate month-level shocks via month fixed effects is typical

---

### 4. Core Analysis 2: DID on the December 2024 Income Requirement Relaxation (Strongest Identification)

This is the point where the policy genuinely changed treatment intensity, so identification is much stronger. The newly eligible group in the 130–200 million won bracket is a natural treatment group.

#### 4.1 DID Model

$$
Birth_{i,t}=\alpha_i+\delta_t+\beta\,(Post^{B}_t \times NewlyEligible_{h(i)})+\Gamma'X_{i,t}+\varepsilon_{i,t}
$$

- Treatment group: $NewlyEligible=1$ (income > 130 million and ≤ 200 million won)
- Recommended comparison group: A group just below the cutoff
  - e.g., $HHInc \in [110\text{ million},130\text{ million}]$ or $[100\text{ million},130\text{ million}]$
    Comparing within similar income ranges reduces confounding.

---

#### 4.2 Advantages of This Design

- Far more persuasive than crude comparisons like top 30% vs. bottom 30%
- The policy shock is clearly pinned to December 2024, making event timing unambiguous
- Considering the fertility lag, effects are likely to appear in the first half of 2025 (policy → conception → birth lag)

---

#### 4.3 Event Study Extension

- Estimate $\beta_k$ relative to December 2024 to examine dynamics through June 2025

---

### 5. Extended Design: Triple-Diff to Further Reduce Confounding

Since the policy is housing-finance in nature, effects are likely greater among non-homeowner, *jeonse*-renting, and housing-unstable groups. If housing status cannot be directly identified from the eligibility-household databases, proxy variables can be constructed (e.g., frequent household moves, young adults, region, workplace enrollment status, etc.).

If possible, introduce a housing-related proxy $Z_i$:

$$
Birth_{i,t}=\alpha_i+\delta_t+\beta\,(Post_t\times Eligible_i) + \theta\,(Post_t\times Eligible_i \times Z_i)+\Gamma'X_{i,t}+\varepsilon_{i,t}
$$

Here, $\theta$ indicates whether the policy effect is larger among groups with greater housing constraints. This substantially increases causal explanatory power (consistency with the mechanism).

---

### 6. Controlling for Marriage Rate, Age Structure, and Economic Conditions: Implementation with This Data

#### 6.1 Marriage-Cohabitation Controls

Implementation order (depending on data availability):

- (If possible) Direct marital status variable: best if marriage-divorce administrative records are available
- (Proxy) Presence of spouse in the same household, couple household composition
- (Proxy) Marriage transition event: the month when a spouse joins the same household

Control approach:

- Version including marital/cohabitation status in $X_{i,t}$
- Alternatively, restrict the sample to the married/cohabiting group (conditional effect) to examine only birth decisions

Caution: Marriage may be on the causal pathway of the policy. Therefore, both sets of results are typically presented in a report:

- Without marriage control: Total effect
- With marriage control: Direct effect (or residual effect) after removing the marriage pathway

---

#### 6.2 Age Structure Controls

With an individual panel, age structure is easy to control.

- Include age (or age group) dummies in $X_{i,t}$
- More aggressively: age × month fixed effects are also possible (if the sample is large)

Recommended: Age-specific analysis

- Split into 25–29, 30–34, 35–39 for separate event studies
  Fertility response is almost always most sensitive in the early 30s.

---

#### 6.3 Economic Conditions Controls

Month fixed effects $\delta_t$ absorb most nationwide common economic shocks. The problem arises when shocks differentially affect income groups.

Possible reinforcements:

- Workplace enrollee status × month interaction (absorbs heterogeneity in income-employment stability shocks)
- If region is available, region × month fixed effects (removes regional housing-economic shocks)
- Insurance premium changes (income shock proxy) included in $X_{i,t}$

---

### 7. Two Implementation Options Reflecting Fertility's Time Lag

#### 7.1 Option A: Based on Birth Timing (Simplest)

Use the event study presented above to directly examine whether births increased several months after the policy.

---

#### 7.2 Option B: Approximate Conception Date Alignment (More Precise)

Approximate the conception month by subtracting 9 months from the birth month and realign policy exposure accordingly.

- Example: $ConceptionMonth = BirthMonth - 9$
  This approach more directly examines whether the policy influenced the conception decision. With large datasets, this method substantially increases persuasiveness.

---

### 8. Sample Design (Recommended)

Since the policy targets newborn/birth households, the following is recommended as the primary analysis:

- Women of childbearing age, 20–44 (too broad reduces signal-to-noise)
- Separate by parity using birth history
  - First birth: strongly linked to marriage and housing constraints
  - Second or higher: linked to income and housing stability
- Center the event study on January 2022 – June 2025 (2 years pre-policy + 18 months post-policy)
  Using the full period from 2006 has advantages, but sufficient length for pre-policy trend verification is all that is needed. Long-term trends can be presented as supplementary material.

---

### 9. Placebo and Robustness Checks (Required to Claim Reliability)

#### 9.1 Fake Policy Timing Placebo

- Run the same DID as if the policy started in January 2022. If an effect appears, the design is not adequately controlling for confounders.

#### 9.2 Varying the Comparison Group

- Slightly adjust the eligibility criterion (e.g., 120 million, 140 million won)
- Verify whether results are similar when changing the bandwidth around the cutoff

#### 9.3 Pre-Treatment Coefficient Verification (Parallel Trends)

If the pre-policy $\beta_k$ values in the event study are systematically different from 0, interpretation should be conservative.
