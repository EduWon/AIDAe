### 5. Interpretation

![Kinematics](https://raw.githubusercontent.com/pilsunchoi/images4/main/22-2.png)

#### 5.1 Horizontal Velocity

The mean horizontal velocity is 194.39 px/s (SD = 137.10), with a coefficient of variation (CV) reaching 70.5%. This indicates substantial acceleration and deceleration as the subject enters and exits the frame. The median (226.76) exceeds the mean, indicating a negatively skewed distribution, which reflects the influence of low-speed segments during entry and exit. In the stable segment (3.0–3.8 s), a relatively constant horizontal velocity is maintained, suggesting an approximation to **steady-state running**.

#### 5.2 Vertical Velocity

The mean vertical velocity is 8.40 px/s, close to zero, but the SD is extremely large at 231.11 with a CV of 2,751.1%. This indicates **strong periodic oscillation** in the vertical direction during running. Positive and negative vertical velocities alternate, causing the mean to cancel out while the variance remains very large — a classic gait/running pattern. The symmetric range of Min (−990.27) and Max (942.92) shows that the ascending phase (takeoff/flight) and descending phase (landing) repeat at similar magnitudes.

#### 5.3 Resultant Speed

The mean resultant speed is 284.32 px/s (SD = 168.32, CV = 59.2%), higher than horizontal velocity due to the contribution of the vertical component. The median (281.08) is nearly identical to the mean, indicating a relatively symmetric distribution. The maximum (990.57) arose from sudden vertical movements (such as landing impact), with the instantaneous peak speed reaching approximately 3.5 times the mean.

#### 5.4 Acceleration

The mean resultant acceleration is 4,442.23 px/s², and the SD of vertical acceleration (7,632) is approximately 3.5 times greater than the SD of horizontal acceleration (2,192). This reflects the fact that **vertical impact accelerations during landing and takeoff are much greater than those in the horizontal direction** during running. The extreme values of vertical acceleration (Min = −29,837, Max = 22,344) can be interpreted as strong deceleration (negative) at landing and acceleration (positive) at takeoff, consistent with vertical ground reaction force (GRF) patterns.

#### 5.5 Head Vertical Oscillation & Gait Cycle

Five peaks and five valleys were observed in the time series of head vertical position. The mean peak-to-peak interval is 0.309 s, corresponding to a stride frequency of approximately **3.2 strides/s**. Considering that running stride frequency typically ranges from 2.5 to 4.0 strides/s, the subject is estimated to have been running at a moderate to fast pace. The detrended pattern of vertical oscillation confirms regular oscillation of approximately ±20–30 px, reflecting the alternation of **flight phase and stance phase** characteristic of running.

#### 5.6 Body Morphology Variables

The variation in body height (Mean = 161.71 px, CV = 34.6%) and width (Mean = 93.76 px, CV = 31.3%) is primarily attributable to two factors. First, during frame entry and exit, the partially visible silhouette reduces measured height and width. Second, **limb opening and closing causes periodic changes in silhouette shape** within the running cycle (height tends to be maximal during stance and decreases during flight). Silhouette area (Mean = 7,672 px², CV = 47.1%) follows a similar pattern, showing relatively consistent values of approximately 9,000–12,000 px² during the stable segment.

### 6. Limitations & Recommendations

**Pixel-unit analysis:** Since measurements were computed in pixel units without spatial calibration (a known reference distance), conversion to absolute physical units such as m/s or m/s² is not possible. Using calibration markers in future work would enable actual velocity and acceleration computation.

**2D analysis:** Because this is a single-camera 2D video, movement in the depth direction (z-axis) is not captured. 3D kinematic analysis requires multi-view capture or depth sensors.

**Silhouette-based tracking:** Since the entire silhouette's center of mass and contour were tracked rather than joint positions, detailed kinematic variables such as joint angles (knee angle, hip angle, etc.) and stride length are not included. Applying a pose estimation model would enable more precise analysis.

**Low-resolution video:** At 352×240 resolution, subtle 1–2 px inter-frame movements may manifest as quantization errors, which may have partially contributed to the high variability in acceleration variables.

**Short effective segment:** The body was detected in only approximately 1.67 seconds of the total 6.31-second video, limiting the statistical stability of gait cycle estimation. Securing a longer capture segment would improve the reliability of cycle analysis.

## **Appendix: Python Code**

```python
"""
=============================================================
 Running Gait Kinematic Analysis
 - Method: Background Subtraction + Contour Analysis
 - Output: Core kinematic variable extraction → Descriptive statistics → Visualization
=============================================================
"""

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib
from scipy.signal import detrend, find_peaks

matplotlib.rcParams['font.family'] = 'DejaVu Sans'
matplotlib.rcParams['axes.unicode_minus'] = False


# =============================================================
# ★ Video file path (modify to match your environment)
# =============================================================
VIDEO_PATH = '../Data/22-1.mp4'


# =============================================================
# PART 1. Load video and check basic information
# =============================================================

cap = cv2.VideoCapture(VIDEO_PATH)
fps = cap.get(cv2.CAP_PROP_FPS)                    # Frame rate
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))      # Horizontal resolution
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))    # Vertical resolution
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))  # Total frame count
dt = 1.0 / fps                                      # Inter-frame time interval

print(f"FPS: {fps:.2f}, Resolution: {width}x{height}, "
      f"Total frames: {total_frames}, Duration: {total_frames/fps:.2f}s")
cap.release()


# =============================================================
# PART 2. Track body silhouette via background subtraction
# =============================================================

# Initialize MOG2 background subtractor
bg_sub = cv2.createBackgroundSubtractorMOG2(
    history=30, varThreshold=50, detectShadows=False
)

# 1st pass: Learn background model from first 30 frames
cap_bg = cv2.VideoCapture(VIDEO_PATH)
for i in range(min(30, total_frames)):
    ret, frame = cap_bg.read()
    if not ret:
        break
    bg_sub.apply(frame)
cap_bg.release()

# 2nd pass: Detect body and extract feature points from all frames
cap = cv2.VideoCapture(VIDEO_PATH)
kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))  # Morphological operation kernel
all_data = []

for frame_idx in range(total_frames):
    ret, frame = cap.read()
    if not ret:
        break

    # Generate foreground mask and remove noise
    fg_mask = bg_sub.apply(frame, learningRate=0.001)
    fg_mask = cv2.morphologyEx(fg_mask, cv2.MORPH_OPEN, kernel)   # Opening (noise removal)
    fg_mask = cv2.morphologyEx(fg_mask, cv2.MORPH_CLOSE, kernel)  # Closing (fill holes)
    fg_mask = cv2.dilate(fg_mask, kernel, iterations=2)           # Dilation (strengthen connections)

    # Contour detection
    contours, _ = cv2.findContours(
        fg_mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE
    )

    row = {'frame': frame_idx, 'time': frame_idx * dt}

    if contours:
        largest = max(contours, key=cv2.contourArea)  # Largest contour = body
        area = cv2.contourArea(largest)

        if area > 500:  # Minimum area threshold
            x, y, w, h = cv2.boundingRect(largest)  # Bounding box

            # Compute centroid
            M = cv2.moments(largest)
            if M['m00'] > 0:
                cx = M['m10'] / M['m00']
                cy = M['m01'] / M['m00']
            else:
                cx, cy = x + w / 2, y + h / 2

            # Extract extreme points (head/feet/left/right)
            topmost = tuple(largest[largest[:, :, 1].argmin()][0])     # Topmost → head
            bottommost = tuple(largest[largest[:, :, 1].argmax()][0])  # Bottommost → feet
            leftmost = tuple(largest[largest[:, :, 0].argmin()][0])    # Leftmost
            rightmost = tuple(largest[largest[:, :, 0].argmax()][0])   # Rightmost

            row.update({
                'detected': True,
                'bbox_x': x, 'bbox_y': y, 'bbox_w': w, 'bbox_h': h,
                'centroid_x': cx, 'centroid_y': cy,
                'area': area,
                'head_x': topmost[0], 'head_y': topmost[1],
                'foot_x': bottommost[0], 'foot_y': bottommost[1],
                'left_x': leftmost[0], 'right_x': rightmost[0],
                'body_height': bottommost[1] - topmost[1],  # Body height
                'body_width': rightmost[0] - leftmost[0],   # Body width
            })
        else:
            row['detected'] = False
    else:
        row['detected'] = False

    all_data.append(row)

cap.release()


# =============================================================
# PART 3. Compute kinematic variables
# =============================================================

df = pd.DataFrame(all_data)
detected = df[df['detected'] == True].copy()

print(f"\nPerson detected in {len(detected)} / {len(df)} frames")
print(f"Detection range: frame {detected['frame'].min()} to {detected['frame'].max()}")
print(f"Time range: {detected['time'].min():.3f}s to {detected['time'].max():.3f}s")

# Displacement (inter-frame position change)
detected['disp_x'] = detected['centroid_x'].diff()
detected['disp_y'] = detected['centroid_y'].diff()

# Velocity (displacement / time interval, px/s)
detected['vel_x'] = detected['disp_x'] / dt   # Horizontal velocity
detected['vel_y'] = detected['disp_y'] / dt   # Vertical velocity
detected['speed'] = np.sqrt(detected['vel_x']**2 + detected['vel_y']**2)  # Resultant speed

# Acceleration (1st derivative of velocity, px/s²)
detected['acc_x'] = detected['vel_x'].diff() / dt   # Horizontal acceleration
detected['acc_y'] = detected['vel_y'].diff() / dt   # Vertical acceleration
detected['acc_mag'] = np.sqrt(detected['acc_x']**2 + detected['acc_y']**2)  # Resultant acceleration

# Head vertical oscillation (for gait cycle analysis)
detected['head_disp_y'] = detected['head_y'].diff()
detected['head_vel_y'] = detected['head_disp_y'] / dt

# Body morphology change
detected['body_width_change'] = detected['body_width'].diff()
detected['height_change'] = detected['body_height'].diff()


# =============================================================
# PART 4. Compute descriptive statistics
# =============================================================

# Remove first-frame NaN from 1st and 2nd derivatives
valid = detected.dropna(subset=['vel_x', 'acc_x']).copy()

# List of kinematic variables for analysis
kinematic_vars = {
    'Horizontal Velocity Vx (px/s)':   valid['vel_x'],
    'Vertical Velocity Vy (px/s)':     valid['vel_y'],
    'Resultant Speed (px/s)':          valid['speed'],
    'Horizontal Accel. ax (px/s2)':    valid['acc_x'],
    'Vertical Accel. ay (px/s2)':      valid['acc_y'],
    'Resultant Accel. |a| (px/s2)':    valid['acc_mag'],
    'Head Vertical Vel. (px/s)':       valid['head_vel_y'].dropna(),
    'Body Height (px)':                valid['body_height'],
    'Body Width (px)':                 valid['body_width'],
    'Silhouette Area (px2)':           valid['area'],
}

# Compute descriptive statistics per variable (mean, SD, min, max, range, median, CV)
stats_rows = []
for name, series in kinematic_vars.items():
    s = series.dropna()
    stats_rows.append({
        'Variable': name,
        'N': len(s),
        'Mean': f"{s.mean():.2f}",
        'SD': f"{s.std():.2f}",
        'Min': f"{s.min():.2f}",
        'Max': f"{s.max():.2f}",
        'Range': f"{s.max() - s.min():.2f}",
        'Median': f"{s.median():.2f}",
        'CV(%)': (
            f"{(s.std() / abs(s.mean()) * 100):.1f}"
            if abs(s.mean()) > 0.01 else "N/A"
        ),
    })

stats_df = pd.DataFrame(stats_rows)

print("\n===== Descriptive Statistics =====\n")
print(stats_df.to_string(index=False))

# --- Global movement summary ---
total_disp_x = detected['centroid_x'].iloc[-1] - detected['centroid_x'].iloc[0]  # Total horizontal displacement
total_disp_y = detected['centroid_y'].iloc[-1] - detected['centroid_y'].iloc[0]  # Total vertical displacement
total_time = detected['time'].iloc[-1] - detected['time'].iloc[0]               # Total movement time
avg_speed = np.sqrt(total_disp_x**2 + total_disp_y**2) / total_time             # Average speed

print(f"\n===== Global Movement Summary =====")
print(f"Total horizontal displacement: {total_disp_x:.1f} px "
      f"({'L to R' if total_disp_x > 0 else 'R to L'})")
print(f"Total vertical displacement: {total_disp_y:.1f} px")
print(f"Total movement time: {total_time:.3f} s")
print(f"Average speed: {avg_speed:.1f} px/s")

# --- Gait cycle analysis (based on head vertical oscillation) ---
head_y = valid['head_y'].values
if len(head_y) > 5:
    peaks, _ = find_peaks(head_y, distance=3)      # Vertical oscillation peaks
    valleys, _ = find_peaks(-head_y, distance=3)    # Vertical oscillation valleys
    print(f"\n===== Gait Cycle Analysis (Head Vertical Oscillation) =====")
    print(f"Number of peaks: {len(peaks)}")
    print(f"Number of valleys: {len(valleys)}")
    if len(peaks) >= 2:
        peak_intervals = np.diff(valid['time'].iloc[peaks].values)  # Time intervals between peaks
        print(f"Mean peak interval (stride period): "
              f"{np.mean(peak_intervals):.3f} s")
        print(f"Estimated stride frequency: "
              f"{1 / np.mean(peak_intervals):.1f} strides/s")


# =============================================================
# PART 5. Visualization (screen display only, no file saving)
# =============================================================

t = valid['time'].values

fig, axes = plt.subplots(4, 2, figsize=(16, 20))
fig.suptitle('Running Gait Kinematic Analysis',
             fontsize=16, fontweight='bold', y=0.98)

# (a) Centroid trajectory
ax = axes[0, 0]
sc = ax.scatter(valid['centroid_x'], valid['centroid_y'],
                c=valid['time'], cmap='viridis', s=30, zorder=3)
ax.plot(valid['centroid_x'], valid['centroid_y'], 'k-', alpha=0.3, linewidth=1)
ax.set_xlabel('X position (px)')
ax.set_ylabel('Y position (px)')
ax.set_title('(a) Centroid Trajectory')
ax.invert_yaxis()
plt.colorbar(sc, ax=ax, label='Time (s)')
ax.grid(True, alpha=0.3)

# (b) Horizontal velocity
ax = axes[0, 1]
ax.plot(t, valid['vel_x'], 'b-', linewidth=1.5, label='Vx')
ax.axhline(valid['vel_x'].mean(), color='r', linestyle='--', alpha=0.7,
           label=f'Mean={valid["vel_x"].mean():.1f}')
ax.fill_between(t,
                valid['vel_x'].mean() - valid['vel_x'].std(),
                valid['vel_x'].mean() + valid['vel_x'].std(),
                alpha=0.15, color='r')  # Mean ± 1SD range
ax.set_xlabel('Time (s)')
ax.set_ylabel('Velocity (px/s)')
ax.set_title('(b) Horizontal Velocity (Vx)')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

# (c) Vertical velocity
ax = axes[1, 0]
ax.plot(t, valid['vel_y'], 'g-', linewidth=1.5, label='Vy')
ax.axhline(0, color='k', linestyle='-', alpha=0.3)       # Zero reference line
ax.axhline(valid['vel_y'].mean(), color='r', linestyle='--', alpha=0.7,
           label=f'Mean={valid["vel_y"].mean():.1f}')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Velocity (px/s)')
ax.set_title('(c) Vertical Velocity (Vy)')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

# (d) Resultant speed
ax = axes[1, 1]
ax.plot(t, valid['speed'], 'm-', linewidth=1.5, label='Speed')
ax.axhline(valid['speed'].mean(), color='r', linestyle='--', alpha=0.7,
           label=f'Mean={valid["speed"].mean():.1f}')
ax.fill_between(t,
                valid['speed'].mean() - valid['speed'].std(),
                valid['speed'].mean() + valid['speed'].std(),
                alpha=0.15, color='r')  # Mean ± 1SD range
ax.set_xlabel('Time (s)')
ax.set_ylabel('Speed (px/s)')
ax.set_title('(d) Resultant Speed')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

# (e) Resultant acceleration magnitude
ax = axes[2, 0]
ax.plot(t, valid['acc_mag'], 'r-', linewidth=1, alpha=0.8, label='|a|')
ax.axhline(valid['acc_mag'].mean(), color='k', linestyle='--', alpha=0.7,
           label=f'Mean={valid["acc_mag"].mean():.0f}')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Acceleration (px/s$^2$)')
ax.set_title('(e) Acceleration Magnitude')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

# (f) Head vertical oscillation (for gait cycle identification)
ax = axes[2, 1]
head_y_vals = valid['head_y'].values
ax.plot(t, head_y_vals, 'b-', linewidth=1.5, label='Head Y')
if len(head_y_vals) > 5:
    detrended = detrend(head_y_vals)  # Remove linear trend
    ax2 = ax.twinx()
    ax2.plot(t, detrended, 'r-', linewidth=1, alpha=0.6, label='Detrended')
    ax2.set_ylabel('Detrended (px)', color='r')
    ax2.tick_params(axis='y', labelcolor='r')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Head Y position (px)')
ax.set_title('(f) Head Vertical Oscillation')
ax.invert_yaxis()  # Image coordinate system: Y-axis positive downward
ax.grid(True, alpha=0.3)

# (g) Body height & width changes
ax = axes[3, 0]
ax.plot(t, valid['body_height'], 'b-', linewidth=1.5, label='Height')
ax.plot(t, valid['body_width'], 'r-', linewidth=1.5, label='Width')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Size (px)')
ax.set_title('(g) Body Height & Width')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

# (h) Silhouette area (body shape change indicator)
ax = axes[3, 1]
ax.plot(t, valid['area'], 'purple', linewidth=1.5, label='Area')
ax.axhline(valid['area'].mean(), color='r', linestyle='--', alpha=0.7,
           label=f'Mean={valid["area"].mean():.0f}')
ax.set_xlabel('Time (s)')
ax.set_ylabel('Area (px$^2$)')
ax.set_title('(h) Silhouette Area')
ax.legend(fontsize=9)
ax.grid(True, alpha=0.3)

plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()


# =============================================================
# PART 6. Statistics table image (screen display only)
# =============================================================

fig2, ax_tbl = plt.subplots(figsize=(18, 6))
ax_tbl.axis('off')

col_labels = stats_df.columns.tolist()
cell_text = stats_df.values.tolist()

# Create table
table = ax_tbl.table(
    cellText=cell_text, colLabels=col_labels,
    loc='center', cellLoc='center'
)
table.auto_set_font_size(False)
table.set_fontsize(8)
table.scale(1, 1.6)

# Header style (dark background + white text)
for j in range(len(col_labels)):
    table[0, j].set_facecolor('#2C3E50')
    table[0, j].set_text_props(color='white', fontweight='bold')

# Alternating row background for data rows
for i in range(1, len(cell_text) + 1):
    for j in range(len(col_labels)):
        table[i, j].set_facecolor('#ECF0F1' if i % 2 == 0 else 'white')

fig2.suptitle('Descriptive Statistics of Kinematic Variables',
              fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\n===== Analysis Complete =====")
```
