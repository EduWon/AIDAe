## <span style="color: blue;">**4. Introduction to World Bank WDI**</span>

## **Claude: Explanation**

The World Bank WDI (World Development Indicators) is a comprehensive collection of global development data covering more than 200 countries and territories with over 1,500 indicators. The API provides access to approximately 16,000 time-series indicators, many of which go back more than 50 years.

WDI is currently organized into six major thematic areas, and the World Bank Open Data site classifies them into 20 detailed topics. Below is a summary organized by major area.

---

### 1. Poverty & Inequality
Indicators measuring the incidence and depth of poverty according to national and international definitions, as well as economic inequality in income and wealth within and across countries.

Key indicator codes include poverty headcount ratio (`SI.POV.DDAY` — percentage of population living on less than $2.15/day), Gini coefficient (`SI.POV.GINI`), and income distribution.

---

### 2. People
A broad set of indicators covering education, health, nutrition, mortality, employment and unemployment, social protection, demographics and migration, and gender — providing a comprehensive picture of social progress.

Detailed topics under World Bank Open Data include Education, Health, Gender, Social Development, and Social Protection & Labor.

Key indicators include total population (`SP.POP.TOTL`), life expectancy (`SP.DYN.LE00.IN`), birth rate (`SP.DYN.CBRT.IN`), infant mortality rate (`SP.DYN.IMRT.IN`), primary school enrollment rate (`SE.PRM.ENRR`), literacy rate (`SE.ADT.LITR.ZS`), and unemployment rate (`SL.UEM.TOTL.ZS`).

---

### 3. Environment
Includes indicators on the use of natural resources such as water and energy, pollution, deforestation, and habitat loss, used to assess the extent of climate change and humanity's impact on the planet.

Detailed topics include Climate Change, Energy & Mining, and Environment.

Key indicators include CO₂ emissions (`EN.ATM.CO2E.KT`), access to electricity (`EG.ELC.ACCS.ZS`), renewable energy consumption share (`EG.FEC.RNEW.ZS`), forest area as percentage of land (`AG.LND.FRST.ZS`), and energy use per capita (`EG.USE.PCAP.KG.OE`).

---

### 4. Economy
Indicators measuring the structure and growth of the world economy, including national accounts indicators such as GDP, GNI, value added, and capital formation, as well as balance of payments, finance, consumption, and adjusted net savings.

Detailed topics include Economy & Growth, External Debt, and Financial Sector.

Key indicators include GDP (`NY.GDP.MKTP.CD`), GDP growth rate (`NY.GDP.MKTP.KD.ZG`), GNI per capita (`NY.GNP.PCAP.CD`), consumer price inflation (`FP.CPI.TOTL.ZG`), and current account balance (`BN.CAB.XOKA.CD`). Financial indicators such as foreign direct investment, deposit interest rate, lending interest rate, real interest rate, domestic credit, official exchange rate, and total foreign reserves are also included.

---

### 5. States & Markets
Indicators on private investment, the public sector, the financial system, telecommunications and transportation infrastructure, and science and technology — providing a picture of business environments, government functions, and the diffusion of new technologies across countries.

Detailed topics include Infrastructure, Private Sector, Public Sector, and Science & Technology.

Key indicators include commercial bank branches (per 100,000 adults), number of listed domestic companies, stock market capitalization (as % of GDP), stock turnover ratio, as well as military expenditure (`MS.MIL.XPND.GD.ZS`), internet users as percentage of population (`IT.NET.USER.ZS`), and R&D expenditure (`GB.XPD.RSDV.GD.ZS`).

---

### 6. Global Links
Indicators providing an overview of the scale and direction of economic flows — trade, remittances, equity, and debt — as well as tourism and migration, covering the processes, structures, and partnerships that enable economies to thrive.

Detailed topics include Trade, Aid Effectiveness, and Urban Development.

Key indicators include merchandise exports (`BX.GSR.MRCH.CD`), trade as percentage of GDP (`NE.TRD.GNFS.ZS`), personal remittances received, net migration, and ODA received (`DT.ODA.ODAT.CD`).

---

### API Call Format

All indicators can be accessed using the following format:

```
https://api.worldbank.org/v2/country/{country_code}/indicator/{indicator_code}?date={start}:{end}&format=json&per_page=20000
```

Access is available without an API key or any other authentication method, so data can be retrieved as long as you know the desired indicator code. The full list of indicators can be queried at `https://api.worldbank.org/v2/indicator?format=json&per_page=20000`.

## 💻 **Python Code**

````python
"""
GDP Growth Analysis of Major Countries Before and After COVID-19 via World Bank WDI API
Indicator: NY.GDP.MKTP.KD.ZG (GDP growth, annual %)
"""

import pandas as pd
import requests
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import numpy as np
from tabulate import tabulate


# ══════════════════════════════════════════════════
# 1. World Bank API data fetching function
# ══════════════════════════════════════════════════

def wb_fetch(countries, indicator, start=2000, end=2024):
    """Fetch indicator data from the World Bank WDI API v2"""
    cc = ";".join(countries)
    url = f"https://api.worldbank.org/v2/country/{cc}/indicator/{indicator}"
    params = {
        "date": f"{start}:{end}",
        "format": "json",
        "per_page": 20000
    }
    r = requests.get(url, params=params, timeout=60)
    r.raise_for_status()
    data = r.json()[1]  # [0]=paging, [1]=records
    rows = []
    for x in data:
        rows.append({
            "country": x["country"]["value"],
            "iso3": x["countryiso3code"],
            "year": int(x["date"]),
            "value": x["value"]
        })
    df = pd.DataFrame(rows).dropna(subset=["value"])
    df = df.sort_values(["iso3", "year"])
    return df


# ══════════════════════════════════════════════════
# 2. Target countries and period settings
# ══════════════════════════════════════════════════

# Major advanced economies + China (G7 + South Korea, China, Australia)
countries = ["KOR", "USA", "JPN", "CHN", "DEU", "GBR", "FRA", "CAN", "ITA", "AUS"]
start, end = 2017, 2023

# Country code to display name mapping
COUNTRY_NAMES = {
    "KOR": "South Korea",  "USA": "United States", "JPN": "Japan",
    "CHN": "China",        "DEU": "Germany",       "GBR": "United Kingdom",
    "FRA": "France",       "CAN": "Canada",        "ITA": "Italy",
    "AUS": "Australia",
}

# Country display order (used in graphs/tables)
DISPLAY_ORDER = list(COUNTRY_NAMES.values())


# ══════════════════════════════════════════════════
# 3. API call and data collection
# ══════════════════════════════════════════════════

print("Fetching GDP growth data from the World Bank WDI API...")
gdp = wb_fetch(countries, "NY.GDP.MKTP.KD.ZG", start, end)

# Convert country names to concise display names
gdp["country"] = gdp["iso3"].map(COUNTRY_NAMES)
print(f"Successfully received {len(gdp)} data records.\n")


# ══════════════════════════════════════════════════
# 4. Create pivot table and console output
# ══════════════════════════════════════════════════

# Pivot: rows=country, columns=year, values=GDP growth rate
pivot = gdp.pivot_table(index="country", columns="year", values="value")
pivot = pivot.reindex(DISPLAY_ORDER)

# COVID-19 shock (2020 - 2019)
pivot["Shock\n(2020-2019)"] = (pivot[2020] - pivot[2019]).round(1)
# Recovery rebound (2021 - 2020)
pivot["Recovery\n(2021-2020)"] = (pivot[2021] - pivot[2020]).round(1)

# Add average row
avg = pivot.mean().round(1)
avg.name = "[Average]"
table_display = pd.concat([pivot, avg.to_frame().T])

print("=" * 110)
print("  GDP Growth Rate (annual %) \u2014 World Bank WDI (NY.GDP.MKTP.KD.ZG)")
print("=" * 110)
print(tabulate(table_display, headers="keys", tablefmt="fancy_grid", floatfmt=".1f"))
print()


# ══════════════════════════════════════════════════
# 5. Generate graphs
# ══════════════════════════════════════════════════

plt.rcParams.update({
    "font.family": "DejaVu Sans",
    "axes.unicode_minus": False,
    "figure.dpi": 150,
})

# Pivot for plotting (years only, no extra columns)
plot_pivot = gdp.pivot_table(index="country", columns="year", values="value")
plot_pivot = plot_pivot.reindex(DISPLAY_ORDER)

years = sorted(plot_pivot.columns.tolist())

# Color palette
COLORS = [
    "#E63946", "#1D3557", "#457B9D", "#E9C46A", "#2A9D8F",
    "#264653", "#F4A261", "#6A4C93", "#1982C4", "#8AC926",
]

# Highlighted countries (thicker lines + circle markers)
HIGHLIGHT = {"South Korea", "United States", "China", "Japan"}

fig = plt.figure(figsize=(20, 18), facecolor="white")
fig.suptitle(
    "GDP Growth Around COVID-19 Pandemic\n"
    "Source: World Bank WDI (NY.GDP.MKTP.KD.ZG)",
    fontsize=16, fontweight="bold", y=0.98,
)


# -- Chart 1: Line chart (overall trends) --
ax1 = fig.add_subplot(3, 1, 1)

for i, (country, row) in enumerate(plot_pivot.iterrows()):
    is_highlight = country in HIGHLIGHT
    ax1.plot(
        years, row.values,
        label=country,
        color=COLORS[i],
        linewidth=2.8 if is_highlight else 1.4,
        marker="o" if is_highlight else "s",
        markersize=7 if is_highlight else 4,
        alpha=1.0 if is_highlight else 0.65,
        zorder=3 if is_highlight else 2,
    )

# Highlight COVID-19 onset year with shaded region
ax1.axvspan(2019.5, 2020.5, color="#FFD6D6", alpha=0.4)
ax1.annotate(
    "COVID-19\nPandemic", xy=(2020, plot_pivot.max().max() * 0.85),
    fontsize=11, color="#C0392B", fontweight="bold", ha="center",
)
# Zero baseline
ax1.axhline(y=0, color="gray", linewidth=0.8, linestyle="--", alpha=0.7)

ax1.set_title("GDP Growth Rate Trends (2017\u20132023)", fontsize=14, fontweight="bold", pad=12)
ax1.set_xlabel("Year", fontsize=11)
ax1.set_ylabel("GDP Growth (%)", fontsize=11)
ax1.set_xticks(years)
ax1.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.1f%%"))
ax1.legend(loc="lower left", ncol=5, fontsize=8.5, framealpha=0.9)
ax1.grid(True, alpha=0.3)


# -- Chart 2: Heatmap (country x year growth rates) --
ax2 = fig.add_subplot(3, 1, 2)

data_arr = plot_pivot.values
im = ax2.imshow(data_arr, cmap="RdYlGn", aspect="auto", vmin=-11, vmax=9)

# Display numeric labels in each cell
for i in range(data_arr.shape[0]):
    for j in range(data_arr.shape[1]):
        val = data_arr[i, j]
        if np.isnan(val):
            continue
        txt_color = "white" if abs(val) > 6 else "black"
        ax2.text(j, i, f"{val:.1f}%", ha="center", va="center",
                 fontsize=9, fontweight="bold", color=txt_color)

ax2.set_xticks(range(len(years)))
ax2.set_xticklabels(years, fontsize=10)
ax2.set_yticks(range(len(plot_pivot.index)))
ax2.set_yticklabels(plot_pivot.index, fontsize=10)
ax2.set_title("Heatmap of GDP Growth by Country and Year", fontsize=14, fontweight="bold", pad=12)

cbar = fig.colorbar(im, ax=ax2, shrink=0.8, pad=0.02)
cbar.set_label("GDP Growth (%)", fontsize=10)


# -- Chart 3: COVID-19 shock vs recovery bar chart --
ax3 = fig.add_subplot(3, 1, 3)

country_labels = plot_pivot.index.tolist()
x = np.arange(len(country_labels))
width = 0.35

# 2020 growth (shock) and 2021 growth (recovery)
shock_vals = plot_pivot[2020].values
recovery_vals = plot_pivot[2021].values

bars1 = ax3.bar(x - width / 2, shock_vals, width,
                label="2020 (COVID Shock)", color="#E63946",
                edgecolor="white", linewidth=0.5)
bars2 = ax3.bar(x + width / 2, recovery_vals, width,
                label="2021 (Recovery)", color="#2A9D8F",
                edgecolor="white", linewidth=0.5)

# Display numeric labels above/below bars
for bar in bars1:
    h = bar.get_height()
    y_pos = h - 0.3 if h < 0 else h + 0.2
    ax3.text(bar.get_x() + bar.get_width() / 2, y_pos,
             f"{h:.1f}%", ha="center", va="bottom" if h >= 0 else "top",
             fontsize=8, fontweight="bold", color="#C0392B")
for bar in bars2:
    h = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width() / 2, h + 0.2,
             f"{h:.1f}%", ha="center", va="bottom",
             fontsize=8, fontweight="bold", color="#1B7A6B")

ax3.axhline(y=0, color="gray", linewidth=0.8, linestyle="--")
ax3.set_xticks(x)
ax3.set_xticklabels(country_labels, fontsize=9, rotation=25, ha="right")
ax3.set_ylabel("GDP Growth (%)", fontsize=11)
ax3.set_title("COVID-19 Shock (2020) vs Recovery (2021)", fontsize=14, fontweight="bold", pad=12)
ax3.legend(fontsize=10)
ax3.grid(axis="y", alpha=0.3)
ax3.yaxis.set_major_formatter(mticker.FormatStrFormatter("%.1f%%"))

plt.show()


# ══════════════════════════════════════════════════
# 6. Summary of key findings
# ══════════════════════════════════════════════════

print("-- Key Findings --")
min_2020_idx = plot_pivot[2020].idxmin()
print(f"  * Largest decline in 2020: {min_2020_idx} ({plot_pivot.loc[min_2020_idx, 2020]:.1f}%)")

# Countries with positive growth in 2020
positive_2020 = plot_pivot[plot_pivot[2020] > 0][2020]
if not positive_2020.empty:
    for c, v in positive_2020.items():
        print(f"  * Maintained positive growth in 2020: {c} (+{v:.1f}%)")

max_2021_idx = plot_pivot[2021].idxmax()
print(f"  * Largest rebound in 2021: {max_2021_idx} (+{plot_pivot.loc[max_2021_idx, 2021]:.1f}%)")

shock = plot_pivot[2020] - plot_pivot[2019]
print(f"  * Largest COVID-19 shock (2020-2019 gap): {shock.idxmin()} ({shock.min():.1f}%p)")
````
