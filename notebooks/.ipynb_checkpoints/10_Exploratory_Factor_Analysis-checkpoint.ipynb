## <span style="color: blue;">**2. Exploratory Factor Analysis**</span>

## **Gemini's Analysis**

### 1. What Is Exploratory Factor Analysis?

Exploratory Factor Analysis (EFA) is a statistical technique that analyzes the correlations among numerous variables (here, 14 service satisfaction items) and groups them into a smaller number of latent factors. It is primarily used for dimension reduction and identifying the underlying structure of data.

* **Purpose:** Simplify complex data and identify unobserved fundamental causes (factors) to facilitate interpretation.
* **Principle:** Variables with high inter-correlations are likely to belong to the same factor.
* **Key concepts:**
  * **Factor Loading:** Indicates the degree of correlation between each variable and a factor; a larger absolute value implies stronger association with that factor.
  * **Rotation:** Rotates the factor axes so that variables load more clearly onto specific factors, improving interpretability. The most commonly used method is orthogonal rotation (Varimax).

---

## <span style="color: blue;">**3. What Is Exploratory Factor Analysis?**</span>

## **ChatGPT's Explanation**

### 1) What Is Factor Analysis? (Purpose)

Factor Analysis is **a method for summarizing many variables into a few 'hidden common factors.'**

The core idea is:

* If multiple variables **move similarly together**,
* There is a common **unseen cause (latent factor)** behind them.

For example:

* Seat comfort
* Cabin cleanliness
* Food satisfaction

If these values tend to increase together
→ We can interpret this as a single factor called "in-flight comfort."

The purposes are:

1. **Dimension reduction**: Organize many variables into a few factors
2. **Structure understanding**: Discover hidden structure in data
3. **Index construction**: Combine multiple items into a single score (e.g., satisfaction index)

---

### 2) When Is It Used?

Factor analysis is especially common in:

* Survey data (satisfaction, attitudes, psychology)
* Service quality evaluation
* Customer experience analysis
* Psychology, marketing, and social sciences

The condition is straightforward:

* Variables must be **correlated with each other**.
* If they are independent, there is no reason to group them into factors.

---

### 3) Basic Concept (Intuition)

Factor analysis assumes the following structure:

Each variable = common factor + unique factor

In equation form:

$$
X_j = \lambda_{j1}F_1 + \lambda_{j2}F_2 + \cdots + \epsilon_j
$$

* $F$: Common factors (latent factors)
* $\lambda$: Factor loadings (degree of association)
* $\epsilon$: Unique component of that variable (error)

Interpretation is simple:

* Large loading → Well represents that factor’s character
* e.g., Seat comfort loading 0.8 on "comfort factor"
  → Seat comfort is a variable that well explains the comfort dimension

---

### 4) Analysis Procedure (Methodological Flow)

Factor analysis typically proceeds through the following steps.

#### (1) Check Correlation Structure

Verify whether correlations exist among variables.

---

#### (2) Determine Number of Factors

Common criteria:

* Eigenvalue > 1 (Kaiser's criterion)
* Scree plot (elbow point)
* Parallel analysis (most recommended)

---

#### (3) Factor Extraction

Common methods:

* Principal factor method
* Maximum Likelihood

At this stage, only the "directions of the factors" are identified.

---

#### (4) Rotation

The most important step.

Purpose:
→ **Make interpretation easier**

Common rotations:

* Varimax: Assumes independence between factors (most widely used)
* Oblimin: Allows correlations between factors

Before rotation: All variables load modestly on all factors
After rotation: Each variable loads heavily on only specific factors
→ Structure becomes clearer

---

#### (5) Interpretation

For example:

| Variable      | F1  | F2  |
| ------------- | --- | --- |
| Seat comfort  | 0.8 | 0.1 |
| Food          | 0.7 | 0.2 |
| Crew courtesy | 0.1 | 0.8 |

→ F1 = "Cabin Environment"
→ F2 = "Service"

---

### 5) What You Get from the Results

Factor analysis yields:

1. **Factor structure**

   * The hidden dimensions of the data

2. **Factor scores**

   * Each observation’s "latent trait score"
   * Can be used subsequently in regression, clustering, etc.

Example:

* Each customer’s "service satisfaction factor score"

---

### 6) Factor Analysis vs. PCA (Commonly Confused)

Key difference:

* PCA: Aims to explain variance (mathematical dimension reduction)
* Factor Analysis: **Aims to explain latent structure**

In other words:

PCA → Data compression
FA → Meaningful structure discovery

---

### 7) One-Line Summary

Factor analysis is **"a method for discovering the few hidden concepts behind many observed variables."**

## **Reference: Python Code**

```python
# 1. Data loading

import pandas as pd

# Load dataframe
df = pd.read_csv('../Data/train.csv')

# Inspect data
print("Data info:")
print(df.info())
```

```python
# 2. Exploratory Factor Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import FactorAnalysis
from sklearn.preprocessing import StandardScaler

# 1. Data preprocessing
# Select survey item columns
survey_cols = ['Inflight wifi service', 'Departure/Arrival time convenient', 'Ease of Online booking', 
               'Gate location', 'Food and drink', 'Online boarding', 'Seat comfort', 
               'Inflight entertainment', 'On-board service', 'Leg room service', 
               'Baggage handling', 'Checkin service', 'Inflight service', 'Cleanliness']

X = df[survey_cols]

# Handle missing values (0 may mean 'not rated', but treated as a valid response here.
# If 0 represents missing, consider mean imputation or exclusion.
# In this dataset, 0 is sometimes interpreted as 'Not Applicable'.
# For factor analysis, we proceed as-is; standardization is key.)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 2. Determine optimal number of factors (Scree Plot & Eigenvalues)
# Compute eigenvalues of the correlation matrix
corr_matrix = np.corrcoef(X_scaled.T)
eigenvalues = np.linalg.eigvals(corr_matrix)
# Sort in descending order
eigenvalues = np.sort(eigenvalues)[::-1]
print("Eigenvalues (top):", eigenvalues[:10])

# Scree Plot
plt.figure(figsize=(6, 4))
plt.plot(range(1, len(eigenvalues) + 1), eigenvalues, marker='o', linestyle='--')
plt.title('Scree Plot')
plt.xlabel('Number of Factors')
plt.ylabel('Eigenvalue')
plt.axhline(y=1, color='r', linestyle='-')  # Kaiser's criterion
plt.grid()
plt.show()

# Determine number of factors by Kaiser's criterion (Eigenvalue > 1)
n_factors = sum(eigenvalues > 1)
print(f"\nOptimal number of factors (Eigenvalue > 1): {n_factors}")

# 3. Perform Factor Analysis
# Use rotation='varimax' to improve interpretability
fa = FactorAnalysis(n_components=n_factors, rotation='varimax', random_state=42)
fa.fit(X_scaled)

# Factor Loadings
fa_components = pd.DataFrame(fa.components_.T, index=survey_cols, columns=[f'Factor {i+1}' for i in range(n_factors)])
print("\nFactor Loadings:")
print(fa_components)

# Visualization (Heatmap)
plt.figure(figsize=(8, 5))
sns.heatmap(fa_components, annot=True, cmap='RdBu_r', center=0, vmin=-1, vmax=1)
plt.title('Factor Loadings (Varimax Rotation)')
plt.show()
```

```python
import numpy as np
import pandas as pd

service_cols = [
    "Inflight wifi service",
    "Departure/Arrival time convenient",
    "Ease of Online booking",
    "Gate location",
    "Food and drink",
    "Online boarding",
    "Seat comfort",
    "Inflight entertainment",
    "On-board service",
    "Leg room service",
    "Baggage handling",
    "Checkin service",
    "Inflight service",
    "Cleanliness",
]

X = df[service_cols].to_numpy(dtype=float)

# Standardize
Xz = (X - X.mean(axis=0)) / X.std(axis=0, ddof=0)

# Correlation matrix
R = np.corrcoef(Xz, rowvar=False)

# Eigendecomposition
eigvals, eigvecs = np.linalg.eigh(R)
idx = np.argsort(eigvals)[::-1]
eigvals, eigvecs = eigvals[idx], eigvecs[:, idx]

# Number of factors
k = 4

# Factor loadings (principal component factor extraction)
L = eigvecs[:, :k] * np.sqrt(eigvals[:k])

def varimax(Phi, gamma=1.0, q=50, tol=1e-6):
    p, k = Phi.shape
    Rm = np.eye(k)
    d = 0.0
    for _ in range(q):
        d_old = d
        Lam = Phi @ Rm
        u, s, vh = np.linalg.svd(
            Phi.T @ (Lam**3 - (gamma / p) * Lam @ np.diag(np.diag(Lam.T @ Lam)))
        )
        Rm = u @ vh
        d = s.sum()
        if d_old != 0 and d / d_old < 1 + tol:
            break
    return Phi @ Rm

L_rot = varimax(L)

# Sort factors by SS loadings
ss = (L_rot**2).sum(axis=0)
order = np.argsort(ss)[::-1]
L_rot = L_rot[:, order]
ss = ss[order]

loading = pd.DataFrame(L_rot, index=service_cols, columns=[f"F{i+1}" for i in range(k)])

# Align signs for interpretive convenience
for c in loading.columns:
    if loading[c].sum() < 0:
        loading[c] *= -1

print("Eigenvalues (top):", eigvals[:10])
print("\nSS loadings:", ss)
print("\nSS/14:", ss / len(service_cols), "sum:", (ss / len(service_cols)).sum())
print()
print(loading.round(3))
```
