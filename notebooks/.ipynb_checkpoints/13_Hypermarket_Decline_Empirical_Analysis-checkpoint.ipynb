**The analytical framework consists of two approaches.**

First, **Shift-Share Decomposition** — This decomposes changes in hypermarket sales into market effect, channel shift, regulatory effect, and residual (lack of innovation), showing which factor dominates in each period. Over time, channel shift becomes overwhelmingly dominant while the regulatory effect diminishes.

Second, **Simple OLS Regression** (Y = β₀ + β₁·RegulationDummy + β₂·OnlinePenetration + β₃·SingleHouseholdRatio + ε) — Based on t-values, online penetration is the most powerful explanatory variable (|t| = 8.0), while the regulation dummy is significant but relatively small in effect size.

**Overall Contribution Estimates:** Online shift ~52%, changes in consumption structure ~22%, business regulations ~15%, lack of innovation ~11%. In other words, even completely lifting regulations would address only about 1/6 of the causes of decline.

For a more rigorous analysis, I would recommend applying DID (Difference-in-Differences) using regional panel data, or using online penetration as an instrumental variable.
