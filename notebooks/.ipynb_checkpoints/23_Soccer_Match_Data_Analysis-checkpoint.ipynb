**2018 FIFA World Cup Final — France vs Croatia (Match ID: 8658)**

### 1. Overview

This document describes the structure and fields of the 2018 FIFA World Cup Final (France vs Croatia) match data provided in **StatsBomb Open Data** format. The data consists of two JSON files.

| File | Size | Description |
|--------|------|------|
| `16-event_8658.json` | 2,978 events | Detailed records of all events during the match (passes, shots, tackles, fouls, etc.) |
| `16-lineup8658.json` | 45 players | Squad entries for both teams (23 + 22), including player info, positions, and card records |

**Match Result:** France 4 – 2 Croatia | **Formations:** France 4-4-2, Croatia 4-3-3 | **Coordinate System:** 120×80 yards

---

### 2. Event Data

The event file is a JSON array recording all actions during the match in chronological order. Each event consists of common fields and type-specific fields.

#### 2-1. Common Fields

Basic fields included in every event:

| Field | Type | Description |
|--------|------|------|
| `id` | UUID | Unique identifier for each event |
| `index` | Integer | Event sequence number (starting from 1) |
| `period` | Integer | Match half (1: first half, 2: second half) |
| `timestamp` | String | Time within the half (HH:MM:SS.sss) |
| `minute` / `second` | Integer | Minutes/seconds based on total match time |
| `type` | Object | Event type (`id`, `name`) |
| `possession` | Integer | Possession sequence number |
| `possession_team` | Object | Team info for the possession |
| `play_pattern` | Object | Play pattern (open play, set piece, etc.) |
| `team` | Object | Team performing the event |
| `player` | Object | Player performing the event (`id`, `name`) |
| `position` | Object | Player position at that moment |
| `location` | Array [x, y] | Pitch coordinates (120×80 basis) |
| `duration` | Float | Event duration (seconds) |
| `related_events` | Array | List of related event UUIDs |

#### 2-2. Event Type Distribution

Distribution of major event types out of 32 total:

| Event Type | Count | Description |
|-------------|------|------|
| Pass | 846 | Pass events (recipient, length, angle, height, outcome, etc.) |
| Ball Receipt* | 747 | Pass reception events |
| Carry | 617 | Ball carries (start/end coordinates) |
| Pressure | 254 | Pressing events against opponents |
| Ball Recovery | 88 | Ball recovery |
| Duel | 53 | 1v1 contests (aerial/tackle, etc.) |
| Clearance | 37 | Clearances |
| Shot | 23 | Shots (includes xG, outcome, freeze frame) |
| Goal Keeper | 32 | Goalkeeper actions (saves, punches, etc.) |
| Foul Committed | 28 | Fouls |
| Dribble | 28 | Dribble attempts |
| Interception | 17 | Interceptions |
| Others | 208 | Substitutions, injury stoppages, camera, tactical shifts, etc. |


## <span style="color: blue;">**2. Match Flow Analysis**</span>


## **Claude Analysis**

An interactive match flow visualization dashboard has been created. The components are as follows:

> 🔗 **2018 FIFA World Cup Final Match Flow Dashboard** (https://claude.ai/public/artifacts/e804fc63-135d-41c8-95bb-37013f47b65e)

![Worldcup](https://raw.githubusercontent.com/pilsunchoi/images4/main/23-2.png)

**Scoring Timeline** — All 6 goals are placed on the time axis to provide an at-a-glance view of the scoring flow.

**Possession Flow Strip** — 182 possession sequences are visualized by half. Each block's width is proportional to the number of events; hovering reveals detailed possession info (time, passes, shots, xG, play pattern). Goals are shown in gold; shots as white dots.

**Momentum Chart** — Based on a 5-minute moving average, it shows changes in match dominance for both teams. From around the 30th minute of the first half through early second half, Croatia overwhelmingly dominated the match, but France scored 2 goals during a brief window of dominance between the 58th and 65th minutes, which is clearly visible.



## <span style="color: blue;">**3. Player Evaluation**</span>


## **Claude Analysis**

A personal performance analysis dashboard has been created for all 27 players who appeared. Click on each player card to expand detailed analysis.

> 🔗 **2018 FIFA World Cup Final Player Evaluation Dashboard** (https://claude.ai/public/artifacts/b878065b-1feb-4837-8014-bd686662baec)

![Worldcup](https://raw.githubusercontent.com/pilsunchoi/images4/main/23-3.png)

**Radar Chart** — Visualizes percentile rankings relative to starting players across 6 axes (Attack, Passing, Carrying, Pressing, Defense, Volume) on a per-90-minute basis. Pogba shows extreme values in both attack and defense, while Brozović shows extreme values in passing and volume.

**Position Change History** — Displays the `positions` array from lineup data as a timeline. For example, Mbappé moved from Right Midfield → Right Center Forward at the 81st minute, and Griezmann's position also changed at the same time, allowing us to trace Didier Deschamps' tactical substitutions in the second half.

**Key Findings:**
- **Brozović** — 271 touches, 99 passes (87 successful), match-highest. The absolute hub of Croatia's build-up
- **Mbappé** — 6 progressive dribbles, 245yd carry distance, the core of France's counter-attacks. Scored 1 goal from 0.06 xG (clinical finishing)
- **Griezmann** — 31 pressures, highest among France players. Overwhelming defensive contribution in the forward press despite being a forward
- **Kanté** — 19 pressures in 54 minutes played, 1 yellow card. Suggests the substitution was for fitness + card management
