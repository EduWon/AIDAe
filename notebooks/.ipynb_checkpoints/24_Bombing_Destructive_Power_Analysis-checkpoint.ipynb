## <span style="color: blue;">**2. Blast Wave Propagation Pattern Estimation Model**</span>

## **ChatGPT: Explanation**

To "estimate" blast wave propagation patterns, modeling is typically divided into two layers: 1) First-order approximation using free-space blast wave scaling, and 2) A second-order model correcting for anisotropy (directionality), reflection, and shielding caused by urban structures. When only "damage traces" are available, as in before/after satellite imagery, the most practical approach is to generate approximate isobars (equal overpressure contours) from layer 1), then distort them using layer 2).

Below is a ready-to-use "estimation model" framework.

---

### 1) First-Order Approximation: Free-Space Spherical Propagation + Hopkinson–Cranz Scaling

Converting the explosion to TNT equivalent mass $W$ (kg TNT), the scaled distance from the blast center at distance $R$ (m) is:

$$
Z = \frac{R}{W^{1/3}}
$$

Then the incident peak overpressure $\Delta P_i$ is expressed as:

$$
\Delta P_i(R) = f(Z)
$$

in the form of an empirical formula/table. In practice, empirical formulas such as Kingery–Bulmash provide $f(Z)$. The key insight is that "similar $Z$ values yield similar blast wave intensities."

Beyond peak overpressure, the positive phase impulse $I_+$, which is more directly related to structural damage, is also used:

$$
I_+(R) = g(Z)
$$

Thus, the first-order model output consists of two radial profiles $(\Delta P_i(R), I_+(R))$ at each distance $R$.

At this stage, the isobars are perfect circles (cross-sections of spheres). If the damage pattern is approximately circular, this approximation alone provides a reasonable explanation.

---

### 2) Reflection/Face Pressure Correction: Reflected Overpressure at Target Surfaces

When a wave strikes a surface head-on, like a building wall, the actual applied pressure increases due to reflection. Denoting the reflected overpressure as $\Delta P_r$:

$$
\Delta P_r = \Delta P_i \cdot \Phi(\Delta P_i, P_0, \gamma)
$$

where $P_0$ is atmospheric pressure and $\gamma$ is the specific heat ratio (for air). $\Phi$ is the reflection coefficient; while the precise formula is fluid dynamics-based, for practical estimation it suffices to capture the directionality that "reflected pressure is greater than incident."

If the damage centers on "wall destruction/window breakage near the blast center," using $\Delta P_r$ provides a better fit.

---

### 3) Second-Order Urban Model: Propagation Kernel That 'Distorts' the Circle

In reality, buildings modify blast waves through:

* Shielding
* Channeling (flowing along alleyways)
* Reflection/multiple reflections
* Diffraction

This is simply modeled as a "spatial weight field" $A(x)$. The effective overpressure at location $x$ is:

$$
\Delta P_{\text{eff}}(x) = \Delta P_i(R(x)) \cdot A(x)
$$

where $R(x)$ is the distance from the blast center to $x$. $A(x)=1$ means identical to free space, 0.5 means halved attenuation, and 1.5 means amplification due to reflection/focusing.

The question is how to construct $A(x)$; for satellite imagery-based analysis, the following approaches are practical:

#### 3.1 Shielding / Line-of-Sight (LOS) Based Model

If the line of sight from the blast center to location $x$ passes through building masses, attenuation is applied:

$$
A(x) = \exp(-k \cdot L_{\text{occ}}(x))
$$

* $L_{\text{occ}}(x)$: Total length of the blast-center-to-$x$ line segment passing through "opaque obstacles (buildings)"
* $k$: Attenuation intensity parameter

This produces a pattern where damage drops sharply in directions where large buildings serve as shields.

#### 3.2 Road/Alley Channeling Correction

The effect of blast waves traveling farther along open corridors (roads) is modeled as "path cost." The effective distance $R_{\text{eff}}(x)$ from the blast center to $x$ is defined as the shortest path:

$$
R_{\text{eff}}(x) = \min_{\pi: 0 \to x} \int_{\pi} c(s) ds
$$

* Open spaces (roads, vacant lots) have low cost $c(s)$
* Building interiors/thick obstacles have high cost

Then:

$$
\Delta P_{\text{eff}}(x) = f\left(\frac{R_{\text{eff}}(x)}{W^{1/3}}\right)
$$

This naturally produces isobars elongated along roads.

This is a typical implementation using "Eikonal/fast marching" methods (not precise CFD, but produces plausible patterns).

---

### 4) Reverse Parameter Estimation from Damage Observations

What we typically obtain from photos is not "continuous pressure" but "damage grades." For example, at each location $x$, damage grade $D(x)$ can be labeled as:

* 0: No damage
* 1: Minor (glass/partial roof)
* 2: Moderate (partial collapse)
* 3: Severe (total collapse)

(manual labeling or segmentation). Then, with threshold overpressures $\theta_j$ per grade:

$$
P(D(x) \ge j) = \sigma\left(\alpha(\Delta P_{\text{eff}}(x) - \theta_j)\right)
$$

* $\sigma(t) = \frac{1}{1+e^{-t}}$
* $\alpha$: Sharpness of the boundary

This allows fitting the unknowns $W, k, \theta_j, \alpha$ via maximum likelihood estimation (MLE) or Bayesian methods.

In practice, it is most stable to first set $A(x)=1$ and roughly fit $W$, then add urban attenuation parameters like $k$ to match the "directionality."

---

### 5) What This Model Produces: "Blast Wave Propagation Pattern Map"

The final output typically consists of two maps:

1. $\Delta P_{\text{eff}}(x)$ heatmap: Isobars appearing distorted from circular form
2. $P(D(x)\ge j)$ map: Damage probability (or expected damage grade) map

Overlaying these two can explain "why damage extended further in a particular direction" through channeling/shielding.

## <span style="color: blue;">**3. Damage Probability Heatmap**</span>

## **ChatGPT: Analysis**

A damage probability (here, "severe damage" probability) heatmap was created. The pixel difference between the before (left) and after (right) images was used as a damage intensity proxy, and from the estimated blast center $(c_x, c_y)$, the probability decreases in logistic form with respect to distance $r$:

$$
p_{\text{severe}}(r)=\frac{1}{1+\exp\left(\frac{r-r_2}{s}\right)}
$$

where $r_2$ is the intermediate isobar radius (cumulative 75% radius) defined earlier, and $s$ is the transition width (approximately $0.1r_2$). That is, this is a "relative" damage probability map normalized so that $p_{\text{severe}}=0.5$ at $r=r_2$ (not based on absolute kPa values).

**Overlay (on the post-bombing image)**

![Bomb](https://raw.githubusercontent.com/pilsunchoi/images4/main/24-2.png)

If desired, "any damage" can also be drawn in the same manner (defined as reaching 0.5 at the outer radius $r_3$).

## 💻 **Python Code**

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# -------------------------------------------------
# 0. Utility: RGB -> Grayscale
# -------------------------------------------------
def to_gray(a):
    # ITU-R BT.601 luminance coefficients (weighted sum)
    return (0.299*a[...,0] + 0.587*a[...,1] + 0.114*a[...,2]).astype(np.float32)

# -------------------------------------------------
# 1. Utility: Integral image-based box blur (mean filter) - very fast
# -------------------------------------------------
def box_blur_fast(img2d, k=9):
    """
    img2d: (H, W) float32
    k: kernel size (odd recommended)
    Returns: (H, W) float32 blurred result
    """
    if k < 1:
        return img2d.copy()

    pad = k // 2

    # Edge padding (same effect as original code)
    p = np.pad(img2d, ((pad, pad), (pad, pad)), mode="edge")

    # Integral image: computes rectangle sums in O(1) using cumulative sums
    # Uses (H+1, W+1) zero-padded cumulative sum format for convenience
    s = np.pad(p, ((1, 0), (1, 0)), mode="constant").cumsum(axis=0).cumsum(axis=1)

    # Compute k×k window sum at each position in one operation
    # Window top-left (i, j) to bottom-right (i+k-1, j+k-1) sum:
    # sum = S[i+k, j+k] - S[i, j+k] - S[i+k, j] + S[i, j]
    H, W = img2d.shape
    sum_k = (
        s[k:k+H, k:k+W]
        - s[0:H,   k:k+W]
        - s[k:k+H, 0:W]
        + s[0:H,   0:W]
    )

    return (sum_k / (k * k)).astype(np.float32)

# -------------------------------------------------
# 2. Load input image and split left/right (before/after bombing)
# -------------------------------------------------
img_path = "../images/24-1.png"
img = Image.open(img_path).convert("RGB")
arr = np.array(img)

h, w, _ = arr.shape
mid = w // 2
left  = arr[:, :mid, :]   # Before bombing
right = arr[:, mid:, :]   # After bombing

# -------------------------------------------------
# 3. Compute before/after difference (damage proxy) + stabilize with blur
# -------------------------------------------------
gL = to_gray(left)
gR = to_gray(right)
diff = np.abs(gR - gL)

# Replace slow double for-loop with integral image box blur
k = 9
blur = box_blur_fast(diff, k=k)

# -------------------------------------------------
# 4. Use top percentile of change as "severe damage" candidates
#    and estimate blast center (weighted centroid)
# -------------------------------------------------
thr = np.percentile(blur, 99.2)   # Top 0.8%
mask = blur >= thr
weights = blur * mask

ys, xs = np.indices(weights.shape)
W = weights.sum()

# Safety check: if threshold is too high and mask is empty
if W == 0:
    cy, cx = h / 2, (mid) / 2
else:
    cy = (ys * weights).sum() / W
    cx = (xs * weights).sum() / W

# -------------------------------------------------
# 5. Compute radius r from blast center
# -------------------------------------------------
dy = ys - cy
dx = xs - cx
r = np.sqrt(dx*dx + dy*dy)

# -------------------------------------------------
# 6. Define 3 isobar radii (r1, r2, r3) from damage intensity
#    cumulative distribution
# -------------------------------------------------
r_flat = r.ravel()
w_flat = weights.ravel()

order = np.argsort(r_flat)
r_sorted = r_flat[order]
w_sorted = w_flat[order]

cum = np.cumsum(w_sorted)
total = cum[-1] if cum.size else 0.0

levels = [0.50, 0.75, 0.90]
radii = []

if total == 0.0:
    # Safety handling when damage pixels are essentially absent
    r1 = r2 = r3 = 1.0
else:
    for lv in levels:
        idx = np.searchsorted(cum, lv * total)
        idx = min(max(idx, 0), len(r_sorted) - 1)
        radii.append(r_sorted[idx])
    r1, r2, r3 = radii

# -------------------------------------------------
# 7. Distance-based damage probability (relative) model: logistic
# -------------------------------------------------
def logistic(z):
    return 1.0 / (1.0 + np.exp(-z))

# Transition width (~10% of radius) to avoid overly sharp step
s_any = max(1.0, 0.10 * r3)
s_sev = max(1.0, 0.10 * r2)

# Set p_any ≈ 0.5 at r=r3, p_sev ≈ 0.5 at r=r2
p_any = logistic((r3 - r) / s_any)
p_sev = logistic((r2 - r) / s_sev)

p_any = np.clip(p_any, 0, 1)
p_sev = np.clip(p_sev, 0, 1)

# -------------------------------------------------
# 8. Visualization: Overlay severe damage probability heatmap
#    on post-bombing image
# -------------------------------------------------
fig, ax = plt.subplots(figsize=(7, 5))
ax.imshow(right)
hm = ax.imshow(p_sev, alpha=0.55)  # Use default colormap
ax.scatter([cx], [cy], s=60, marker="x", linewidths=2)
ax.set_axis_off()
ax.set_title("Estimated Damage Probability Heatmap (severe)")

cbar = plt.colorbar(hm, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Probability")

fig.tight_layout()
plt.show()
```
