## <span style="color: blue;">**1. Recent Trends in Global Cosmetics Trade**</span>

## **Claude: Analysis**

### Market Size and Growth

The global cosmetics market is estimated at approximately $346.1 billion in 2025 and is projected to reach $526.7 billion by 2033, with a compound annual growth rate (CAGR) of about 5.39%. By product category, skincare accounts for the largest share at 43.4%, and by end consumer, women make up 62.6%.

---

### Shifts in Regional Trade Structure

**Rise of Asia-Pacific:** The Asia-Pacific region has become the largest market as of 2025, accounting for 45.5% of the global cosmetics market. It is also expected to record the highest CAGR of 7.18% during the 2026–2031 period.

**Europe's Export Dominance:** Europe continues to lead the export market, holding 48.4% of global cosmetics exports. France in particular maintains its position as the world's largest beauty exporter. However, recent trends tell a different story. In 2025, French cosmetics exports totaled €22.4 billion, a 0.1% decline year-over-year—marking the first contraction since the 2008 financial crisis (excluding COVID). Exports to the US plunged approximately 19%, largely due to US tariff barriers and a weaker dollar.

---

### The K-Beauty Surge: South Korea Breaks Export Records

The global expansion of Korean cosmetics is the most notable trend.

In 2025, South Korea's cosmetics exports reached an all-time high of $11.43 billion, a 12.3% increase year-over-year. During January–April 2025, South Korea surpassed the United States to become the world's second-largest cosmetics exporter, behind France.

Export destinations have also diversified rapidly. Korean cosmetics were exported to 202 countries in 2025, a significant increase from 172 countries in 2024. Notably, the top export market shifted: the United States overtook China to become the largest destination for Korean cosmetics, with exports reaching $2.2 billion. Meanwhile, exports to China fell 19.2% year-over-year, with China's share dropping sharply from 52.9% in 2021.

By product category, basic cosmetics (skincare) accounted for $8.54 billion, or 74.7% of the total, while the fragrance segment posted the highest growth rate at 46.2% year-over-year.

---

### Tariffs and Trade Barriers: The Impact of US Tariff Policy

Changes in US tariff policy in 2025 have sent shockwaves through global cosmetics trade.

Given that imports account for two-thirds of US beauty industry product costs, the average US tariff rate on beauty products surged from 2.4% in 2024 to a trade-weighted 30%. Starting April 5, 2025, the US imposed a 10% baseline tariff on virtually all imports, including cosmetics. Packaging costs are expected to rise 5–10% by late 2025 through 2026, with products using a high proportion of aluminum facing even greater impact.

Additionally, the US FDA abolished the de minimis exemption for imports under $800 starting July 2025, subjecting all cosmetics imports to full review. This has directly affected K-Beauty brands that rely on small-parcel shipments.

Industry responses have been active. Kolmar Korea completed its second manufacturing facility in the US, positioning itself to serve the US, European, and Latin American markets without tariff burdens. While large companies can leverage their own factories and extensive supply chains to shift production to low-tariff countries, small brands that rely on one or two contract manufacturers face margin erosion due to limited flexibility.

---

### Summary of Key Trade Trends

In summary, the global cosmetics trade is undergoing several structural changes. First, export destinations are diversifying rapidly alongside K-Beauty's global expansion. Second, US protectionist policies are dampening exports from France and Europe while prompting Asian exporters like South Korea to expand local production as a strategic response. Third, technological innovation—including AI-driven R&D, personalized product recommendations, and sustainable packaging—along with tightening regulations (such as the EU's Packaging and Packaging Waste Regulation and fragrance allergen labeling requirements) are reshaping the trade landscape.

## <span style="color: blue;">**4. Introduction to Graph Theory**</span>

## **Claude: Analysis**

Graph theory is a branch of mathematics that studies structures composed of **nodes (vertices)** and **edges (links)**. It is used to abstractly represent and analyze relationships or connections between entities.

### Fundamental Concepts

A graph G is defined as **G = (V, E)**, where V is the set of nodes and E is the set of edges. In the cosmetics trade network constructed earlier, each country is a node and each trade flow between countries is an edge.

There are several major types of graphs. An **undirected graph** has edges without direction, meaning the A-B relationship is symmetric. A **directed graph (digraph)** has directional edges like A→B, suitable for representing asymmetric relationships such as exports and imports. A **weighted graph** assigns numerical values (trade volume, distance, etc.) to edges, and a **bipartite graph** divides nodes into two groups with edges only between groups.

### Key Metrics

Several metrics are used to characterize graph properties. **Degree** is the number of edges connected to a node; in directed graphs, incoming edges (in-degree) and outgoing edges (out-degree) are distinguished. The fact that France had the highest out-degree in our analysis means it exports to the most countries.

**Centrality** measures a node's influence within the network. Key types include degree centrality (based on number of connections), betweenness centrality (a measure of brokerage role), and PageRank (the algorithm Google used for webpage ranking, based on the principle that "a node is more important if it receives connections from other important nodes"). The US having the highest PageRank in our cosmetics trade analysis reflects its role as a hub that all major exporters ship to.

**Density** is the ratio of actual edges to the maximum possible number of edges, indicating how tightly connected the network is.

### Origins and History

Graph theory originated with Euler's **Königsberg Bridge Problem** in 1736. The question—"Can one cross all seven bridges exactly once?"—gave rise to the abstract representation of nodes and edges. Euler proved it was impossible, and this became the starting point of graph theory.

### Applications

Graph theory is applied across a wide range of fields: social network analysis (relationships between people), transportation and logistics route optimization, internet and telecommunications network design, protein interaction networks in biology, and international trade flow analysis as demonstrated in this chapter.

### Relationship to Graphical Models

Note that the term **Graphical Model** also refers specifically to models in statistics and machine learning that represent conditional dependency relationships among random variables using graphs. Bayesian networks (directed acyclic graphs) and Markov random fields (undirected graphs) are representative examples. In this case, nodes represent random variables and edges represent dependencies between variables. The trade network we constructed is a network analysis based on graph theory, which differs in purpose from probabilistic graphical models.

## 💻 **Python Code**

```python
"""

Global Cosmetics Trade Network Graph (Graphical Model)

======================================================

Visualizes the trade network as a directed weighted graph

based on 2025 global cosmetics trade data.



Node: Major exporting/importing countries/regions

Edge: Trade flow (arrow direction = export→import, width = trade volume)

"""



import networkx as nx

import matplotlib.pyplot as plt

import matplotlib.patches as mpatches

import numpy as np

from matplotlib.patches import FancyBboxPatch

from matplotlib import patheffects



# ──────────────────────────────────────────────

# 1. Font settings

# ──────────────────────────────────────────────

plt.rcParams['font.family'] = 'DejaVu Sans'

plt.rcParams['font.size'] = 11

plt.rcParams['axes.unicode_minus'] = False



# ──────────────────────────────────────────────

# 2. Node definitions: Major trading countries/regions

#    - size: Node size based on export volume

#    - type: exporter / importer / both

#    - group: Regional classification (for color mapping)

# ──────────────────────────────────────────────

nodes = {

    "France":       {"size": 2800, "type": "exporter",  "group": "europe"},

    "South Korea":  {"size": 2600, "type": "exporter",  "group": "asia"},

    "USA":          {"size": 2400, "type": "both",      "group": "americas"},

    "Germany":      {"size": 1800, "type": "exporter",  "group": "europe"},

    "Italy":        {"size": 1400, "type": "exporter",  "group": "europe"},

    "Japan":        {"size": 1600, "type": "both",      "group": "asia"},

    "China":        {"size": 2200, "type": "both",      "group": "asia"},

    "UK":           {"size": 1200, "type": "both",      "group": "europe"},

    "India":        {"size": 1000, "type": "importer",  "group": "asia"},

    "UAE":          {"size":  800, "type": "importer",  "group": "mena"},

    "Brazil":       {"size":  900, "type": "importer",  "group": "americas"},

    "Indonesia":    {"size":  750, "type": "importer",  "group": "asia"},

    "Vietnam":      {"size":  650, "type": "importer",  "group": "asia"},

    "Poland":       {"size":  550, "type": "importer",  "group": "europe"},

    "Mexico":       {"size":  600, "type": "importer",  "group": "americas"},

}



# ──────────────────────────────────────────────

# 3. Edge definitions: Trade flows (exporter → importer)

#    weight = Trade volume (in billions USD, relative scale)

# ──────────────────────────────────────────────

edges = [

    # France exports (world No. 1, ~EUR 22.4B total)

    ("France", "USA",       5.5),

    ("France", "China",     3.8),

    ("France", "Germany",   3.2),

    ("France", "UK",        2.8),

    ("France", "Japan",     2.0),

    ("France", "Italy",     1.8),

    ("France", "UAE",       1.5),

    ("France", "Brazil",    1.2),

    # South Korea exports (world No. 2, $11.43B)

    ("South Korea", "USA",       4.8),

    ("South Korea", "China",     4.4),

    ("South Korea", "Japan",     2.4),

    ("South Korea", "UK",        1.0),

    ("South Korea", "Vietnam",   1.2),

    ("South Korea", "Indonesia", 0.9),

    ("South Korea", "UAE",       0.8),

    ("South Korea", "India",     0.7),

    ("South Korea", "Poland",    0.6),

    ("South Korea", "Brazil",    0.5),

    ("South Korea", "Mexico",    0.4),

    # USA exports (world No. 3, ~$10.8B)

    ("USA", "China",     2.2),

    ("USA", "UK",        1.8),

    ("USA", "Brazil",    1.5),

    ("USA", "Japan",     1.2),

    ("USA", "Mexico",    1.4),

    ("USA", "India",     0.8),

    # Germany exports

    ("Germany", "USA",    1.5),

    ("Germany", "France", 1.2),

    ("Germany", "UK",     1.0),

    ("Germany", "Poland", 0.8),

    ("Germany", "China",  0.7),

    # Japan exports

    ("Japan", "China",       2.5),

    ("Japan", "USA",         1.0),

    ("Japan", "South Korea", 0.6),

    ("Japan", "Indonesia",   0.5),

    ("Japan", "India",       0.4),

    # Italy exports

    ("Italy", "USA",     1.2),

    ("Italy", "France",  0.9),

    ("Italy", "Germany", 0.8),

    ("Italy", "UK",      0.6),

    # China exports (reverse export trend increasing)

    ("China", "USA",         1.8),

    ("China", "Japan",       1.0),

    ("China", "South Korea", 0.5),

    ("China", "Indonesia",   0.6),

    ("China", "Vietnam",     0.5),

    ("China", "India",       0.7),

]



# ──────────────────────────────────────────────

# 4. Create graph object

# ──────────────────────────────────────────────

G = nx.DiGraph()

for node, attrs in nodes.items():

    G.add_node(node, **attrs)

for src, dst, w in edges:

    G.add_edge(src, dst, weight=w)



# ──────────────────────────────────────────────

# 5. Layout: Initial positions for geographic intuition + spring

# ──────────────────────────────────────────────

initial_pos = {

    "France":      (-0.6,  0.5),

    "Germany":     (-0.3,  0.7),

    "Italy":       (-0.4,  0.2),

    "UK":          (-0.8,  0.8),

    "Poland":      (-0.1,  0.6),

    "South Korea": ( 0.8,  0.4),

    "Japan":       ( 1.0,  0.6),

    "China":       ( 0.6,  0.6),

    "India":       ( 0.4,  0.0),

    "Indonesia":   ( 0.8, -0.3),

    "Vietnam":     ( 0.7, -0.1),

    "USA":         (-0.8, -0.2),

    "Brazil":      (-0.5, -0.6),

    "Mexico":      (-0.7, -0.5),

    "UAE":         ( 0.1,  0.1),

}

pos = nx.spring_layout(G, pos=initial_pos, k=1.8, iterations=80, seed=42)



# ──────────────────────────────────────────────

# 6. Color and size mapping

# ──────────────────────────────────────────────

group_colors = {

    "europe":   "#3B82F6",   # Europe: Blue

    "asia":     "#EF4444",   # Asia: Red

    "americas": "#F59E0B",   # Americas: Gold

    "mena":     "#10B981",   # Middle East: Green

}

node_colors = [group_colors[nodes[n]["group"]] for n in G.nodes()]

node_sizes = [nodes[n]["size"] for n in G.nodes()]



# Normalize edge widths (range: 0.5 to 5.0)

edge_weights = [G[u][v]["weight"] for u, v in G.edges()]

max_w, min_w = max(edge_weights), min(edge_weights)

edge_widths = [0.5 + 4.5 * (w - min_w) / (max_w - min_w) for w in edge_weights]



# Edge color: Based on exporter's group + semi-transparent

edge_colors = [group_colors[nodes[u]["group"]] + "55" for u, v in G.edges()]



# ──────────────────────────────────────────────

# 7. [Figure 1] Main network graph

# ──────────────────────────────────────────────

fig, ax = plt.subplots(1, 1, figsize=(20, 14), facecolor="#0F172A")

ax.set_facecolor("#0F172A")



fig.suptitle(

    "Global Cosmetics Trade Network (2025)",

    fontsize=26, fontweight="bold", color="white", y=0.97

)

ax.set_title(

    "Directed Weighted Graph  |  Node = Country  |  Edge = Trade Flow (Export \u2192 Import)",

    fontsize=12, color="#94A3B8", pad=15

)



# Edges (curved arrows)

nx.draw_networkx_edges(

    G, pos, ax=ax, width=edge_widths, edge_color=edge_colors,

    arrows=True, arrowsize=15, arrowstyle="-|>",

    connectionstyle="arc3,rad=0.1",

    min_source_margin=25, min_target_margin=25, alpha=0.7,

)



# Node shadows

nx.draw_networkx_nodes(G, pos, ax=ax, node_size=[s*1.15 for s in node_sizes],

                       node_color="#000000", alpha=0.3)

# Node bodies

nx.draw_networkx_nodes(G, pos, ax=ax, node_size=node_sizes,

                       node_color=node_colors, edgecolors="white",

                       linewidths=1.5, alpha=0.9)



# Labels + outline for readability

text_items = nx.draw_networkx_labels(G, pos, ax=ax,

    font_size=10, font_weight="bold", font_color="white")

for _, text in text_items.items():

    text.set_path_effects([

        patheffects.withStroke(linewidth=3, foreground="#0F172A"),

        patheffects.Normal()

    ])



# Display trade value labels for top 12 edges

top_edges = sorted(edges, key=lambda x: x[2], reverse=True)[:12]

for src, dst, w in top_edges:

    x1, y1 = pos[src]

    x2, y2 = pos[dst]

    mx = (x1 + x2) / 2

    my = (y1 + y2) / 2

    dx, dy = x2 - x1, y2 - y1

    mx += -dy * 0.04

    my += dx * 0.04

    ax.annotate(

        f"${w}B", xy=(mx, my), fontsize=7.5, fontweight="bold", color="#E2E8F0",

        ha="center", va="center",

        bbox=dict(boxstyle="round,pad=0.2", facecolor="#1E293B",

                  edgecolor="#475569", alpha=0.85, linewidth=0.5),

    )



# Legend (regional colors)

legend_patches = [

    mpatches.Patch(color="#3B82F6", label="Europe (FR, DE, IT, UK, PL)"),

    mpatches.Patch(color="#EF4444", label="Asia-Pacific (KR, JP, CN, IN, ID, VN)"),

    mpatches.Patch(color="#F59E0B", label="Americas (US, BR, MX)"),

    mpatches.Patch(color="#10B981", label="Middle East (UAE)"),

]

leg = ax.legend(handles=legend_patches, loc="upper left", fontsize=10,

    title="Region", title_fontsize=11, facecolor="#1E293B",

    edgecolor="#475569", labelcolor="white", framealpha=0.9)

leg.get_title().set_color("white")



# Info box (legend panel)

info_ax = fig.add_axes([0.02, 0.05, 0.18, 0.13], facecolor="none")

info_ax.set_xlim(0, 10); info_ax.set_ylim(0, 4); info_ax.axis("off")

bg = FancyBboxPatch((0, 0), 10, 4, boxstyle="round,pad=0.3",

    facecolor="#1E293B", edgecolor="#475569", alpha=0.9, linewidth=0.5)

info_ax.add_patch(bg)

info_ax.text(5, 3.4, "Node Size = Export Volume", ha="center", fontsize=10, fontweight="bold", color="white")

info_ax.text(5, 2.6, "Edge Width = Trade Volume ($B)", ha="center", fontsize=9, color="#94A3B8")

info_ax.text(5, 1.8, "Arrow Direction = Export \u2192 Import", ha="center", fontsize=9, color="#94A3B8")

info_ax.text(5, 0.9, "Top labels show bilateral trade ($B)", ha="center", fontsize=8.5, color="#64748B")



# Network statistics panel (bottom-right)

stats_ax = fig.add_axes([0.78, 0.02, 0.21, 0.22], facecolor="none")

stats_ax.set_xlim(0, 10); stats_ax.set_ylim(0, 8); stats_ax.axis("off")

sbg = FancyBboxPatch((0, 0), 10, 8, boxstyle="round,pad=0.3",

    facecolor="#1E293B", edgecolor="#475569", alpha=0.9, linewidth=0.5)

stats_ax.add_patch(sbg)



# Compute centrality metrics

pagerank = nx.pagerank(G, weight="weight")

betweenness = nx.betweenness_centrality(G, weight="weight")

top_pr = sorted(pagerank.items(), key=lambda x: x[1], reverse=True)[:3]

top_bw = sorted(betweenness.items(), key=lambda x: x[1], reverse=True)[:3]



stats_lines = [

    ("Network Statistics", True, "white", 7.3),

    (f"Nodes: {G.number_of_nodes()}  |  Edges: {G.number_of_edges()}", False, "#94A3B8", 6.4),

    (f"Density: {nx.density(G):.3f}", False, "#94A3B8", 5.6),

    ("Top Hubs (PageRank):", True, "#E2E8F0", 4.6),

    (f"  1. {top_pr[0][0]}  ({top_pr[0][1]:.3f})", False, "#F59E0B", 3.8),

    (f"  2. {top_pr[1][0]}  ({top_pr[1][1]:.3f})", False, "#94A3B8", 3.0),

    (f"  3. {top_pr[2][0]}  ({top_pr[2][1]:.3f})", False, "#94A3B8", 2.2),

    ("Top Bridges (Betweenness):", True, "#E2E8F0", 1.2),

    (f"  1. {top_bw[0][0]}  ({top_bw[0][1]:.3f})", False, "#F59E0B", 0.4),

]

for txt, bold, col, yy in stats_lines:

    stats_ax.text(0.8, yy, txt, fontsize=9 if not bold else 10,

                  fontweight="bold" if bold else "normal", color=col)



ax.margins(0.12); ax.axis("off")

fig.text(0.5, 0.01,

    "Source: Korea Customs, FEBEA, BCG, Fortune BI, Korea Herald (2025-2026)  |  Graphical Model: Directed Weighted Network",

    ha="center", fontsize=9, color="#64748B", style="italic")



plt.show()



# ──────────────────────────────────────────────

# 8. [Figure 2] Centrality analysis (3-panel)

# ──────────────────────────────────────────────

in_degree = dict(G.in_degree(weight="weight"))

out_degree = dict(G.out_degree(weight="weight"))



fig2, axes = plt.subplots(1, 3, figsize=(20, 6), facecolor="#0F172A")



# (a) PageRank centrality

pr_sorted = sorted(pagerank.items(), key=lambda x: x[1], reverse=True)

axes[0].barh([x[0] for x in pr_sorted][::-1], [x[1] for x in pr_sorted][::-1],

    color=[group_colors[nodes[n]["group"]] for n in [x[0] for x in pr_sorted]][::-1],

    edgecolor="white", linewidth=0.5, alpha=0.85)

axes[0].set_title("(a) PageRank Centrality", fontsize=14, fontweight="bold", color="white", pad=12)

axes[0].set_xlabel("PageRank Score", fontsize=10, color="#94A3B8")



# (b) Weighted In-Degree (import volume)

in_sorted = sorted(in_degree.items(), key=lambda x: x[1], reverse=True)

axes[1].barh([x[0] for x in in_sorted][::-1], [x[1] for x in in_sorted][::-1],

    color=[group_colors[nodes[n]["group"]] for n in [x[0] for x in in_sorted]][::-1],

    edgecolor="white", linewidth=0.5, alpha=0.85)

axes[1].set_title("(b) Weighted In-Degree (Import Volume)", fontsize=14, fontweight="bold", color="white", pad=12)

axes[1].set_xlabel("Total Weighted In-Degree ($B)", fontsize=10, color="#94A3B8")



# (c) Weighted Out-Degree (export volume)

out_sorted = sorted(out_degree.items(), key=lambda x: x[1], reverse=True)

axes[2].barh([x[0] for x in out_sorted][::-1], [x[1] for x in out_sorted][::-1],

    color=[group_colors[nodes[n]["group"]] for n in [x[0] for x in out_sorted]][::-1],

    edgecolor="white", linewidth=0.5, alpha=0.85)

axes[2].set_title("(c) Weighted Out-Degree (Export Volume)", fontsize=14, fontweight="bold", color="white", pad=12)

axes[2].set_xlabel("Total Weighted Out-Degree ($B)", fontsize=10, color="#94A3B8")



# Unify axis styles

for a in axes:

    a.set_facecolor("#0F172A")

    a.tick_params(colors="#94A3B8")

    for spine in a.spines.values():

        spine.set_color("#334155")



fig2.suptitle("Centrality Analysis of Cosmetics Trade Network",

    fontsize=18, fontweight="bold", color="white", y=1.02)

plt.tight_layout()

plt.show()



# ──────────────────────────────────────────────

# 9. [Figure 3] Korea's export diversification + exporter competition

# ──────────────────────────────────────────────

fig3, (ax3a, ax3b) = plt.subplots(1, 2, figsize=(18, 7), facecolor="#0F172A")



# (a) Shift in Korea's export destination shares (2021 vs 2025)

cats = ["China", "USA", "Japan", "Europe", "SE Asia", "Others"]

v21 = [52.9, 10.2, 8.1, 5.5, 6.0, 17.3]

v25 = [17.6, 19.7, 9.5, 11.0, 8.5, 33.7]

x = np.arange(len(cats)); w = 0.35



b1 = ax3a.bar(x - w/2, v21, w, label="2021", color="#EF4444", alpha=0.7, edgecolor="white", linewidth=0.5)

b2 = ax3a.bar(x + w/2, v25, w, label="2025", color="#F59E0B", alpha=0.9, edgecolor="white", linewidth=0.5)

for bar_set, col in [(b1, "#EF4444"), (b2, "#F59E0B")]:

    for bar in bar_set:

        h = bar.get_height()

        ax3a.text(bar.get_x() + bar.get_width()/2, h + 0.8, f"{h}%",

                  ha="center", fontsize=8.5, color=col, fontweight="bold")



ax3a.set_title("(a) S. Korea Export Share Shift: 2021 vs 2025", fontsize=14, fontweight="bold", color="white", pad=12)

ax3a.set_ylabel("Share (%)", fontsize=11, color="#94A3B8")

ax3a.set_xticks(x); ax3a.set_xticklabels(cats, fontsize=10)

ax3a.set_ylim(0, 62); ax3a.set_facecolor("#0F172A"); ax3a.tick_params(colors="#94A3B8")

ax3a.legend(fontsize=10, facecolor="#1E293B", edgecolor="#475569", labelcolor="white")

for spine in ax3a.spines.values(): spine.set_color("#334155")



# (b) Top 3 exporters: Competition trend (2020-2025)

years = [2020, 2021, 2022, 2023, 2024, 2025]

ax3b.plot(years, [17.5, 19.2, 21.5, 23.3, 22.5, 22.4], "o-", color="#3B82F6", lw=2.5, ms=8, label="France")

ax3b.plot(years, [5.2, 6.8, 7.5, 8.5, 10.3, 11.4],    "s-", color="#EF4444", lw=2.5, ms=8, label="South Korea")

ax3b.plot(years, [9.5, 10.1, 10.8, 11.2, 11.0, 10.8],  "D-", color="#F59E0B", lw=2.5, ms=8, label="USA")



# Key event annotations

ax3b.annotate("KR surpasses US\n(Jan-Apr 2025)", xy=(2025, 11.4), xytext=(2023.3, 15),

    fontsize=9, fontweight="bold", color="#EF4444",

    arrowprops=dict(arrowstyle="->", color="#EF4444", lw=1.5),

    bbox=dict(boxstyle="round,pad=0.3", facecolor="#1E293B", edgecolor="#EF4444", alpha=0.9))

ax3b.annotate("FR first decline\nsince 2008", xy=(2025, 22.4), xytext=(2023, 18.5),

    fontsize=9, fontweight="bold", color="#3B82F6",

    arrowprops=dict(arrowstyle="->", color="#3B82F6", lw=1.5),

    bbox=dict(boxstyle="round,pad=0.3", facecolor="#1E293B", edgecolor="#3B82F6", alpha=0.9))



ax3b.set_title("(b) Top 3 Exporters: Competition Trend (2020-2025)", fontsize=14, fontweight="bold", color="white", pad=12)

ax3b.set_ylabel("Exports ($B)", fontsize=11, color="#94A3B8")

ax3b.set_xlabel("Year", fontsize=11, color="#94A3B8")

ax3b.set_facecolor("#0F172A"); ax3b.tick_params(colors="#94A3B8")

ax3b.legend(fontsize=10, facecolor="#1E293B", edgecolor="#475569", labelcolor="white")

ax3b.grid(axis="y", color="#1E293B", linewidth=0.5)

for spine in ax3b.spines.values(): spine.set_color("#334155")



fig3.suptitle("Cosmetics Trade Dynamics: Export Diversification & Competition",

    fontsize=18, fontweight="bold", color="white", y=1.02)

plt.tight_layout()

plt.show()



# ──────────────────────────────────────────────

# 10. [Figure 4] Tariff impact chart

# ──────────────────────────────────────────────

fig4, ax4 = plt.subplots(figsize=(14, 7), facecolor="#0F172A")

ax4.set_facecolor("#0F172A")



# Tariff rate data (2024 average vs 2025 applied)

tariff_countries = ["China", "Cambodia", "Vietnam", "Switzerland", "EU/France", "S. Korea", "UK", "Baseline"]

tariff_2024 = [2.4, 2.4, 2.4, 2.4, 2.4, 2.4, 2.4, 2.4]

tariff_2025 = [54, 49, 46, 31, 20, 15, 10, 10]



y = np.arange(len(tariff_countries))

h = 0.35

bars_old = ax4.barh(y + h/2, tariff_2024, h, label="2024 Avg. Tariff", color="#334155", edgecolor="white", linewidth=0.5)

bars_new = ax4.barh(y - h/2, tariff_2025, h, label="2025 New Tariff", edgecolor="white", linewidth=0.5,

    color=["#DC2626" if t >= 40 else "#F59E0B" if t >= 15 else "#10B981" for t in tariff_2025])



# Value labels

for bar, val in zip(bars_new, tariff_2025):

    ax4.text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2, f"{val}%",

             va="center", fontsize=11, fontweight="bold", color="white")

for bar, val in zip(bars_old, tariff_2024):

    ax4.text(bar.get_width() + 0.5, bar.get_y() + bar.get_height()/2, f"{val}%",

             va="center", fontsize=9, color="#64748B")



# Display increase multipliers

for i, (old, new) in enumerate(zip(tariff_2024, tariff_2025)):

    mult = new / old

    ax4.text(58, y[i], f"\u00d7{mult:.0f}", fontsize=10, fontweight="bold",

             color="#EF4444" if mult >= 15 else "#F59E0B", va="center")



ax4.set_yticks(y); ax4.set_yticklabels(tariff_countries, fontsize=12)

ax4.set_xlabel("Tariff Rate (%)", fontsize=12, color="#94A3B8")

ax4.set_title("US Import Tariff Shock on Cosmetics (2024 \u2192 2025)",

    fontsize=16, fontweight="bold", color="white", pad=15)

ax4.legend(fontsize=10, facecolor="#1E293B", edgecolor="#475569", labelcolor="white", loc="lower right")

ax4.tick_params(colors="#94A3B8")

ax4.set_xlim(0, 65)

for spine in ax4.spines.values(): spine.set_color("#334155")



# Severity legend annotation

ax4.text(35, -1.2, "\u26a0 Red = severe impact (>40%)  |  Yellow = moderate (15-39%)  |  Green = manageable (<15%)",

         fontsize=9, color="#64748B", ha="center")



plt.tight_layout()

plt.show()
```
