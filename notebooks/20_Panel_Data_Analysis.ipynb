### 4. Hausman Test (FE vs RE)

**Test idea**

- Null hypothesis $H_0$: Individual effects and explanatory variables are independent, so RE is a consistent estimator
- Alternative hypothesis $H_1$: Not independent, so RE is inconsistent and FE is safer

The test is calculated using the common coefficients between FE and RE (`exper`, `expersq`, `union`, `married`).

**Results**

**(Clustered covariance)**
- Chi-square statistic = 16.1453
- Degrees of freedom df = 4
- p-value = 0.00283023

**(Non-robust covariance)**
- Chi-square statistic = 31.5736
- Degrees of freedom df = 4
- p-value = 2.33805e-06

**Conclusion**

- Since the p-value is very small, $H_0$ is rejected
- In this specification, it is natural to conclude that the RE independence assumption is violated, and it is practically reasonable to prefer FE.

---

### 5. One-Line Summary

- Pooled OLS cannot control for time-invariant individual characteristics, so it may overestimate coefficients such as `union` and `married`
- Moving to FE reveals the pure within-individual effect after removing individual-specific factors, and the Hausman test also supports FE

## <span style="color: blue;">**4. Intuitive Understanding of Panel Data Models**</span>

## **ChatGPT: Explanation**

Panel data consists of repeated observations of the same individuals (or firms, regions, etc.) across multiple time periods. The core issue is this:

> Each individual has invisible inherent characteristics (ability, diligence, disposition, etc.) that may be correlated with the explanatory variables. The model differs depending on how these "hidden individual differences" are handled.

Consider the following example:

* Objective: Analyzing factors that affect wages
* Explanatory variables: Education, experience, union membership, etc.
* Problem: Some people are inherently more capable. This capability is unobserved but affects both wages and may also be related to education level.


Let us denote this unobservable individual characteristic as $\alpha_i$. The overall structure can be thought of as:

$$
y_{it} = \beta X_{it} + \alpha_i + u_{it}
$$

where:

* $i$: individual
* $t$: time
* $\alpha_i$: individual-specific characteristic (does not change over time)

How $\alpha_i$ is treated is the key distinction.

---

### 1. Pooled OLS

**Idea**

Pool all data together and run standard OLS.

That is:

* Ignore individual distinctions
* Ignore time distinctions
* Just a simple regression with more observations

**Intuition**

This is equivalent to assuming "everyone starts from the same baseline."

**Problem**

If individual ability $\alpha_i$:

* Affects wages, and
* Is also related to variables like education,

→ Explanatory variables and the error term are correlated
→ OLS bias occurs

**Example**

Higher-ability individuals tend to get more education and earn higher wages.
OLS may overestimate the education effect.

**When to use?**

* When individual differences are unimportant
* Or when one believes individual effects are independent of explanatory variables

---

### 2. Fixed Effects Model (FE)

**Idea**

Allow each individual to have their own unique baseline.

That is:

* Acknowledge that the average level differs across individuals
* But assume that difference does not change over time

**Intuition**

"Do not compare across individuals; only look at changes within the same individual."

**Example**

A's wage change vs. A's experience change
B's wage change vs. B's experience change

→ Fixed characteristics such as individual ability are automatically removed.

Mathematically, this is done by subtracting individual means:

$$
(y_{it} - \bar{y}_i) = \beta (X_{it} - \bar{X}_i) + (u_{it} - \bar{u}_i)
$$

Here, $\alpha_i$ disappears.

**Advantages**

* No problem even if unobserved individual characteristics are correlated with explanatory variables
* Safer for causal interpretation

**Disadvantages**

* Cannot estimate the effect of time-invariant variables
  (e.g., if sex, race, or education do not change, their effects cannot be estimated)

---

### 3. Random Effects Model (RE)

**Idea**

Instead of removing individual effects, treat them as a "random variable."

$$
\alpha_i \sim \text{random}
$$

**Key Assumption**

$$
\alpha_i \text{ and } X_{it} \text{ are independent}
$$

**Intuition**

"Differences exist across individuals, but those differences are pure random noise unrelated to the explanatory variables."

If this assumption is correct:

* Between-individual information + within-individual information can both be used
  → More efficient than FE (smaller standard errors)

**Advantages**

* Can estimate time-invariant variables (e.g., sex, education)
* Efficient

**Disadvantages**

* If $\alpha_i$ is correlated with explanatory variables
  → Severe bias occurs

---

### 4. Intuitive Difference Between FE and RE

**FE's perspective**

"Each individual has hidden characteristics that may be related to the explanatory variables. Let's safely remove them."

**RE's perspective**

"Differences exist across individuals, but they are random and unrelated to the explanatory variables. No need to remove them."

**Key Question**

Is the individual effect $\alpha_i$ correlated with the explanatory variables?

---

### 5. Hausman Test

This is a method to statistically test this question.

**Idea**

* FE is always a consistent estimator (weaker assumptions)
* RE is more efficient if its assumptions hold
* If the assumptions are violated, RE is biased

Therefore:

If RE is correct
→ FE and RE results should be similar

**Test logic**

Null hypothesis $H_0$:
RE is valid
(Individual effects and explanatory variables are independent)

Alternative hypothesis $H_1$:
RE is invalid
(FE should be used)

**Results interpretation**

* Small p-value → FE and RE differ substantially → RE assumption violated → Choose FE
* Large p-value → No significant difference → RE can be used

**Intuition**

A test that asks: "Is it okay to use RE?"

---

### 6. All-at-Once Summary (Intuition)

**Pooled OLS**
→ Ignores individual differences

**Fixed Effects**
→ Each individual has unique characteristics, and these may be related to X → Remove them

**Random Effects**
→ Individual differences exist but are random effects unrelated to X → Use them efficiently

**Hausman**
→ Tests whether RE's independence assumption holds
→ If it fails, use FE

---

As a practical rule of thumb:

* If the goal is causal interpretation with panel data → Think FE first
* If the effect of time-invariant variables is needed → Consider RE
* Then make the final judgment with the Hausman test

## 💻 **Python Code**

```python
# =========================================
# Common setup: Load data + Set panel index
# =========================================
import pandas as pd
import numpy as np
import statsmodels.api as sm
from linearmodels.panel import PooledOLS, PanelOLS, RandomEffects
from linearmodels.datasets import wage_panel

# 1) Load data
data = wage_panel.load()
df = data.copy()

# 2) Set panel index: (individual ID=nr, year=year)
df = df.set_index(["nr", "year"]).sort_index()

# 3) (Optional) To view only numeric columns:
# num_cols = df.select_dtypes(include=[np.number]).columns
# df[num_cols].head()
```

```python
# =========================================
# (1) Variable summary statistics
# =========================================

# Identifiers (nr, year) are already in the index, not in columns.
# Generate summary statistics for numeric variables suitable for regression (may include missing values).
# (Optionally, select specific variables for summary_cols.)

summary_cols = df.columns  # Summarize all columns
# summary_cols = ["lwage", "educ", "exper", "expersq", "union", "married", "black", "hisp", "hours"]  # Example

summary = df[summary_cols].describe().T  # count, mean, std, min, 25%, 50%, 75%, max

# Display
summary
```

```python
# =========================================
# (2) Pooled OLS / Fixed Effects / Random Effects estimation (revised)
# =========================================

import pandas as pd
import numpy as np
import statsmodels.api as sm
from linearmodels.panel import PooledOLS, PanelOLS, RandomEffects

# Specify dependent/explanatory variables
y = df["lwage"]

X_cols = ["educ", "exper", "expersq", "union", "married", "black", "hisp"]
X = df[X_cols]

# Add constant (Pooled/RE typically include a constant)
X_const = sm.add_constant(X)

# -------------------------
# 2-1) Pooled OLS
# -------------------------
pooled_mod = PooledOLS(y, X_const)
pooled_res = pooled_mod.fit(cov_type="clustered", cluster_entity=True)  # Clustered SE by entity

# -------------------------
# 2-2) Fixed Effects (entity FE) - Method A (recommended): auto-drop absorbed variables
# -------------------------
# drop_absorbed=True automatically removes time-invariant variables like educ/black/hisp.
fe_mod = PanelOLS(y, X, entity_effects=True, drop_absorbed=True)
fe_res = fe_mod.fit(cov_type="clustered", cluster_entity=True)

# -------------------------
# 2-3) Random Effects
# -------------------------
re_mod = RandomEffects(y, X_const)
re_res = re_mod.fit(cov_type="clustered", cluster_entity=True)

# -------------------------
# Organize results in table format
# -------------------------
def tidy_table(res):
    out = pd.DataFrame({
        "coef": res.params,
        "std_err": res.std_errors,
        "t": res.tstats,
        "p": res.pvalues
    })
    return out

tbl_pooled = tidy_table(pooled_res)
tbl_fe = tidy_table(fe_res)
tbl_re = tidy_table(re_res)

print("\n# Pooled OLS\n")
print(tbl_pooled.to_markdown())

print("\n# Fixed Effects (drop_absorbed=True)\n")
print(tbl_fe.to_markdown())

print("\n# Random Effects\n")
print(tbl_re.to_markdown())

# Check which variables remain in FE (estimated coefficients)
print("\nCoefficients remaining in FE:", list(fe_res.params.index))
```

```python
# =========================================
# (3) Hausman Test (FE vs RE)
# =========================================
# Concept:
# - H0: Individual effects and explanatory variables are independent -> RE is consistent
# - H1: Not independent -> RE is inconsistent, prefer FE
#
# Calculation (standard form):
#   diff = b_FE - b_RE
#   V = Var(b_FE) - Var(b_RE)
#   H = diff' * V^{-1} * diff  ~ chi2(k)
#
# Note:
# - With clustered (robust) covariance, V may be non-invertible/non-positive-definite.
# - In such cases, recalculating with non-robust covariance is a common approach.
# =========================================

import numpy.linalg as la
from scipy import stats

def hausman_test(fe_res, re_res, common_only=True):
    # 1) Identify common coefficient names
    b_fe = fe_res.params
    b_re = re_res.params

    if common_only:
        common = b_fe.index.intersection(b_re.index)
    else:
        common = b_fe.index.intersection(b_re.index)

    # 2) Extract common coefficients only
    b_fe_c = b_fe.loc[common]
    b_re_c = b_re.loc[common]

    # 3) Extract common portion from covariance matrices
    V_fe = fe_res.cov.loc[common, common]
    V_re = re_res.cov.loc[common, common]

    # 4) Calculate Hausman statistic
    diff = (b_fe_c - b_re_c).to_numpy()
    V = (V_fe - V_re).to_numpy()

    # Use inverse (or pseudo-inverse)
    try:
        V_inv = la.inv(V)
    except la.LinAlgError:
        V_inv = la.pinv(V)

    stat = float(diff.T @ V_inv @ diff)
    df_h = len(common)
    pval = 1 - stats.chi2.cdf(stat, df_h)

    return {
        "stat": stat,
        "df": df_h,
        "pvalue": pval,
        "common_params": list(common)
    }

# 1) First attempt Hausman with clustered SE results
h = hausman_test(fe_res, re_res, common_only=True)

print("Hausman test (clustered cov) results")
print(h)

# 2) (Optional) If clustered cov results seem unstable,
#    refit with non-robust (default) covariance and recalculate Hausman.
#    Example below.

fe_res_nr = fe_mod.fit()  # Default (non-robust) covariance
re_res_nr = re_mod.fit()  # Default (non-robust) covariance

h_nr = hausman_test(fe_res_nr, re_res_nr, common_only=True)

print("\nHausman test (non-robust cov) results")
print(h_nr)

# 3) Pretty-print as markdown table:
hausman_tbl = pd.DataFrame(
    {"chi2": [h["stat"]], "df": [h["df"]], "pvalue": [h["pvalue"]]}
)
print("\n# Hausman (clustered)\n")
print(hausman_tbl.to_markdown(index=False))

hausman_tbl_nr = pd.DataFrame(
    {"chi2": [h_nr["stat"]], "df": [h_nr["df"]], "pvalue": [h_nr["pvalue"]]}
)
print("\n# Hausman (non-robust)\n")
print(hausman_tbl_nr.to_markdown(index=False))
```
